Xue et al. (2022) model the fracture surface as a random height field
$$
a(x,y),
$$
defined on a planar domain $(x,y)$. The surface is assumed to be self-affine. For two points separated by a vector $(\Delta x,\Delta y)$, the height increment is defined as
$$
h(x,y;\Delta x,\Delta y)
=
a(x+\Delta x,\,y+\Delta y) - a(x,\,y).
$$
The self-affine assumption is expressed through the statistics of these increments. First, the increments have zero mean,
$$
\Big\langle a(x+\Delta x,\,y+\Delta y) - a(x,\,y) \Big\rangle = 0,
$$
so that the surface exhibits no systematic upward or downward trend in a statistical sense. Second, the variance of the increments depends on the separation distance only and follows a power-law scaling. Let
$$
r = \sqrt{(\Delta x)^2 + (\Delta y)^2}
$$
denote the separation distance. The variance of the height increments at distance $r$ is defined by
$$
\sigma^2(r)
=
\Big\langle 
\big[a(x+\Delta x,\,y+\Delta y) - a(x,\,y)\big]^2 
\Big\rangle.
$$
For a self-affine surface with Hurst exponent $H$, Xue et al. assume
$$
\sigma^2(r) = r^{2H}\,\sigma^2(1),
$$
or, equivalently,
$$
\sigma(r) \propto r^{H},
$$
where $\sigma^2(1)$ is the variance at unit separation. Thus, the typical amplitude of the height difference between two points grows with distance $r$ as a power law with exponent $H$. In practice, a modified Successive Random Addition (SRA) algorithm is employed to construct a discrete field $a_{ij} = a(x_i,y_j)$ whose increments are approximately Gaussian, have zero mean, and satisfy the variance scaling law above.

To quantify the roughness of the generated three-dimensional fracture surface in terms of the traditional Joint Roughness Coefficient (JRC), Xue et al. use the slope-based parameter $Z_2$ introduced by Tse and Cruden (1979). One-dimensional profiles are extracted from the surface along the flow direction. For a given profile, sampled at points $x_i$ with corresponding elevations $\delta_i$ $(i=0,\dots,N_t)$, the parameter $Z_2$ is defined as
$$
Z_2
=
\left[
\frac{1}{N_t}
\sum_{i=1}^{N_t}
\left(
\frac{\delta_i - \delta_{i-1}}{x_i - x_{i-1}}
\right)^2
\right]^{1/2}.
$$
This quantity represents the root-mean-square of the local slope along the profile. The JRC associated with that profile is then obtained from the empirical correlation
$$
\mathrm{JRC}
=
32.2 + 32.47 \,\log_{10}\!\big(Z_2\big).
$$
For a three-dimensional surface, Xue et al. extract multiple (e.g. 20) parallel profiles, compute $Z_2$ and JRC for each, and define a representative roughness value as the arithmetic mean of the individual JRC values,
$$
\overline{\mathrm{JRC}}
=
\frac{1}{N_p}
\sum_{k=1}^{N_p} \mathrm{JRC}^{(k)},
$$
where $N_p$ is the number of profiles and $\mathrm{JRC}^{(k)}$ is the JRC computed from the $k$-th profile. In this way, the self-affine surface generated by the SRA-type procedure is characterized a posteriori by an effective JRC consistent with conventional joint roughness descriptions.


```python
import os
import numpy as np
import vtk
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # needed for 3D plotting


# ---------------------------------------------------------------------------
# Helper: compute sigma_{delta h} from a 3D surface (height field a(x,y))
# ---------------------------------------------------------------------------

def compute_sigma_delta_h(surface, dx, dy):
    """
    Compute the standard deviation of height differences sigma_{delta h} for a
    2D surface a(x,y), using forward differences in x and y and averaging.
    """
    dz_dx = surface[1:, :] - surface[:-1, :]
    dz_dy = surface[:, 1:] - surface[:, :-1]

    sigma_x = np.std(dz_dx)
    sigma_y = np.std(dz_dy)

    # RMS combination of the two directions
    sigma_dh = np.sqrt(0.5 * (sigma_x**2 + sigma_y**2))
    return sigma_dh


# ---------------------------------------------------------------------------
# Invert Tse–Cruden to get target sigma_{delta h} from target JRC
# ---------------------------------------------------------------------------

def sigma_dh_from_target_jrc(JRC_target, dx):
    """
    For a desired JRC_target (Tse–Cruden), estimate the corresponding
    sigma_{delta h} at grid spacing dx by:
      JRC = 32.2 + 32.47 log10(Z2)
      Z2 = sqrt(<slope^2>) ~ sigma_dh / dx
    => sigma_dh_target = Z2_target * dx
    """
    Z2_target = 10 ** ((JRC_target - 32.2) / 32.47)
    sigma_dh_target = Z2_target * dx
    return sigma_dh_target


# ---------------------------------------------------------------------------
# 1) Xue-style 3D self-affine surface: SRA / midpoint-displacement
# ---------------------------------------------------------------------------

def generate_surface_xue_sra_3d(nx, ny, Lx, Ly, H, target_sigma_dh):
    """
    Generate a 3D self-affine fracture surface a(x,y) using a 2D
    diamond–square (midpoint-displacement) algorithm (SRA-style).
    """
    x = np.linspace(0.0, Lx, nx)
    y = np.linspace(0.0, Ly, ny)
    surface = np.zeros((nx, ny), dtype=float)

    # Initialize corners
    surface[0, 0]   = 0.0
    surface[0, -1]  = 0.0
    surface[-1, 0]  = 0.0
    surface[-1, -1] = 0.0

    size_x = nx - 1
    size_y = ny - 1
    step = 2 ** int(np.floor(np.log2(min(size_x, size_y))))

    while step > 1:
        half = step // 2
        seg_length = (step / (nx - 1)) * Lx  # approximate physical length in x
        sigma_level = seg_length ** H

        # Diamond step
        for i in range(0, nx - 1, step):
            for j in range(0, ny - 1, step):
                i0, j0 = i, j
                i1, j1 = i + step, j + step
                ic, jc = i + half, j + half

                c = 0.25 * (
                    surface[i0, j0] + surface[i1, j0] +
                    surface[i0, j1] + surface[i1, j1]
                )
                surface[ic, jc] = c + np.random.normal(scale=sigma_level)

        # Square step: horizontal midpoints
        for i in range(0, nx - 1, step):
            for j in range(0, ny, step):
                ic = i + half
                jc = j
                if ic >= nx:
                    continue
                vals = [surface[i, jc]]
                if i + step < nx:
                    vals.append(surface[i+step, jc])
                if jc - half >= 0:
                    vals.append(surface[ic, jc-half])
                if jc + half < ny:
                    vals.append(surface[ic, jc+half])
                c = np.mean(vals)
                if surface[ic, jc] == 0.0:
                    surface[ic, jc] = c + np.random.normal(scale=sigma_level)

        # Square step: vertical midpoints
        for i in range(0, nx, step):
            for j in range(0, ny - 1, step):
                ic = i
                jc = j + half
                if jc >= ny:
                    continue
                vals = [surface[ic, j]]
                if j + step < ny:
                    vals.append(surface[ic, j+step])
                if ic - half >= 0:
                    vals.append(surface[ic-half, jc])
                if ic + half < nx:
                    vals.append(surface[ic+half, jc])
                c = np.mean(vals)
                if surface[ic, jc] == 0.0:
                    surface[ic, jc] = c + np.random.normal(scale=sigma_level)

        step = half

    # Remove global mean and scale
    surface -= np.mean(surface)
    dx = Lx / (nx - 1)
    dy = Ly / (ny - 1)
    sigma_current = compute_sigma_delta_h(surface, dx, dy)
    if sigma_current > 0.0:
        surface *= (target_sigma_dh / sigma_current)

    return x, y, surface


# ---------------------------------------------------------------------------
# 2) JRC helpers (Xue/Tse–Cruden style)
# ---------------------------------------------------------------------------

def compute_z2_profile(x, z):
    dx = x[1:] - x[:-1]
    dz = z[1:] - z[:-1]
    slopes2 = (dz / dx)**2
    Z2 = np.sqrt(np.mean(slopes2))
    return Z2

def compute_jrc_from_z2(Z2):
    return 32.2 + 32.47 * np.log10(Z2)

def estimate_jrc_xue_style(surface, x, y, n_profiles=20, direction='x'):
    nx, ny = surface.shape
    jrc_list = []

    if direction == 'x':
        indices = np.linspace(0, ny - 1, n_profiles, dtype=int)
        for j in indices:
            z_profile = surface[:, j]
            Z2 = compute_z2_profile(x, z_profile)
            jrc_list.append(compute_jrc_from_z2(Z2))
    else:
        indices = np.linspace(0, nx - 1, n_profiles, dtype=int)
        for i in indices:
            z_profile = surface[i, :]
            Z2 = compute_z2_profile(y, z_profile)
            jrc_list.append(compute_jrc_from_z2(Z2))

    jrc_values = np.array(jrc_list)
    return float(np.mean(jrc_values)), jrc_values


# ---------------------------------------------------------------------------
# 3) VTK export
# ---------------------------------------------------------------------------

def export_surface_to_vtk_structured(x, y, surface, filename, scalar_name="height"):
    nx, ny = surface.shape
    grid = vtk.vtkStructuredGrid()
    grid.SetDimensions(nx, ny, 1)

    points = vtk.vtkPoints()
    points.SetNumberOfPoints(nx * ny)

    values = vtk.vtkDoubleArray()
    values.SetName(scalar_name)
    values.SetNumberOfValues(nx * ny)

    pid = 0
    for i in range(nx):
        for j in range(ny):
            px = x[i]
            py = y[j]
            pz = surface[i, j]
            points.SetPoint(pid, px, py, pz)
            values.SetValue(pid, surface[i, j])
            pid += 1

    grid.SetPoints(points)
    grid.GetPointData().SetScalars(values)

    writer = vtk.vtkStructuredGridWriter()
    writer.SetFileName(filename)
    writer.SetInputData(grid)
    writer.Write()
    print(f"✅ VTK written: {filename}")


# ---------------------------------------------------------------------------
# Example usage + Jupyter visualization
# ---------------------------------------------------------------------------

if __name__ == "__main__":
    # Output folder
    out_dir = "_out"
    os.makedirs(out_dir, exist_ok=True)

    # Domain and grid
    Lx, Ly = 1.0, 1.0
    nx, ny = 129, 129         # 2^k+1 is good for Xue-style
    H = 0.8

    dx = Lx / (nx - 1)
    dy = Ly / (ny - 1)

    # --- Choose a target JRC (engineering sense) ---
    JRC_target = 20.0  # you can change this to e.g. 4, 12, 16, ...
    target_sigma_dh = sigma_dh_from_target_jrc(JRC_target, dx)
    print(f"Target JRC           : {JRC_target}")
    print(f"Target sigma_delta_h : {target_sigma_dh:.6e} m")

    # -------------------------------
    # Xue-style surface
    # -------------------------------
    x, y, surf_xue = generate_surface_xue_sra_3d(nx, ny, Lx, Ly, H, target_sigma_dh)
    sigma_dh_xue = compute_sigma_delta_h(surf_xue, dx, dy)
    jrc_xue_mean, jrc_xue_profiles = estimate_jrc_xue_style(
        surf_xue, x, y, n_profiles=20, direction='x'
    )
    print("\nXue-style surface:")
    print(f"  sigma_delta_h [m]   ≈ {sigma_dh_xue:.6e}")
    print(f"  mean JRC (from Z2)  ≈ {jrc_xue_mean:.2f}")

    export_surface_to_vtk_structured(
        x, y, surf_xue,
        filename=os.path.join(out_dir, "surface_xue.vtk"),
        scalar_name="height"
    )

    # -------------------------------
    # Save JRC summary to _out
    # -------------------------------
    summary_path = os.path.join(out_dir, "roughness_JRC_summary.txt")
    with open(summary_path, "w") as f:
        f.write("Roughness & JRC summary (Xue-style only)\n")
        f.write("=======================================\n\n")
        f.write(f"Target JRC            : {JRC_target:.2f}\n")
        f.write(f"Target sigma_delta_h  : {target_sigma_dh:.6e} m\n\n")

        f.write("Xue-style surface:\n")
        f.write(f"  sigma_delta_h [m]   : {sigma_dh_xue:.6e}\n")
        f.write(f"  mean JRC (from Z2)  : {jrc_xue_mean:.4f}\n")

    print(f"\n✅ JRC summary written: {summary_path}")

    # -------------------------------
    # Jupyter visualization (matplotlib 3D)
    # -------------------------------
    X, Y = np.meshgrid(x, y, indexing='ij')

    fig = plt.figure(figsize=(6, 5))
    ax1 = fig.add_subplot(1, 1, 1, projection='3d')
    ax1.plot_surface(X, Y, surf_xue, linewidth=0, antialiased=False)
    ax1.set_title(f"Xue-style surface\nJRC ≈ {jrc_xue_mean:.1f}")
    ax1.set_xlabel("x [m]")
    ax1.set_ylabel("y [m]")
    ax1.set_zlabel("z [m]")

    plt.tight_layout()
    plt.show()
```

In this study, the fracture mechanical aperture is decomposed as (cf.\ Xue et al., 2022; Barton and Bandis, 1982)


\begin{equation}
  b = b_0 - \Delta b_n(\sigma_n) + \Delta b_{\text{dil}},
\end{equation}


where $b_0$ is the initial mechanical aperture, $\Delta b_n(\sigma_n)$ is the normal closure induced by the effective normal stress $\sigma_n$, and $\Delta b_{\text{dil}}$ is the dilatant opening.

The normal closure is described by a Bandis--Barton-type hyperbolic law,


\begin{equation}
  \Delta b_n(\sigma_n)
  = \frac{\sigma_n}{\sigma_n + k_n \, \Delta b_m}\,\Delta b_m ,
\end{equation}


with $k_n$ the normal stiffness and $\Delta b_m$ the maximum possible closure (Barton and Bandis, 1982).

In a general case, shear-induced dilatancy is related to shear slip $u_s$ and dilation angle $\psi$ via


\begin{equation}
  \Delta b_{\text{dil}} = u_s \, \tan\psi .
\end{equation}


However, in the present numerical demo no explicit shear slip $u_s$ is applied. Instead, $\Delta b_{\text{dil}}$ is prescribed directly as a uniform dilatant increment in the third loading stage. Consequently, the mean aperture evolution is


\begin{equation}
  b_1 = b_0, \qquad
  b_2 = b_0 - \Delta b_n(\sigma_n), \qquad
  b_3 = b_0 - \Delta b_n(\sigma_n) + \Delta b_{\text{dil}} ,
\end{equation}


which is consistent with the Xue-style decomposition $b = b_0 - \Delta b_n + \Delta b_s$, with $\Delta b_s \equiv \Delta b_{\text{dil}}$ imposed rather than generated by finite shear slip.


```python
import os
import numpy as np
import matplotlib.pyplot as plt
import vtk

# ============================================================
# 1. MECHANICAL LAW: hyperbolic normal closure
# ============================================================

def normal_closure_hyperbolic(sigma_n, k_n, delta_bm):
    r"""
    Hyperbolic normal-closure law (Bandis/Barton type):

        Δb_n(σ_n) = [σ_n / (σ_n + k_n Δb_m)] * Δb_m

    Parameters
    ----------
    sigma_n : float
        Effective normal stress [Pa], compression positive.
    k_n : float
        Normal stiffness [Pa/m].
    delta_bm : float
        Maximum possible closure [m].

    Returns
    -------
    delta_bn : float
        Normal closure [m]. Δb_n > 0 means the fracture closes.
    """
    return (sigma_n / (sigma_n + k_n * delta_bm)) * delta_bm


# ============================================================
# 1b. DILATION FROM SHEAR SLIP
# ============================================================

def dilation_from_slip(u_s, psi_deg):
    r"""
    Shear-induced dilatancy from slip and dilation angle:

        Δb_dil = u_s · tan(ψ)

    where
      u_s     : shear slip along the joint [m]
      ψ       : dilation angle [deg]
      Δb_dil  : normal opening due to dilatancy [m]
    """
    psi_rad = np.deg2rad(psi_deg)
    return u_s * np.tan(psi_rad)


# ============================================================
# 2. COMPUTE THREE APERTURES: b0 → closed → dilated
# ============================================================

def compute_aperture_stages(b0, delta_bm, k_n, sigma_n, u_s, psi_deg):
    r"""
    Compute three mean mechanical apertures:

      Step 1: b1 = b0
      Step 2: b2 = b0 - Δb_n(σ_n)
      Step 3: b3 = b2 + Δb_dil

    where
      Δb_n(σ_n)  : hyperbolic normal closure
      Δb_dil     : shear-induced dilatancy = u_s · tan(ψ)

    Parameters
    ----------
    b0 : float
        Initial mechanical aperture [m].
    delta_bm : float
        Maximum possible normal closure [m].
    k_n : float
        Normal stiffness [Pa/m].
    sigma_n : float
        Effective normal stress [Pa], compression positive.
    u_s : float
        Imposed shear slip along the joint [m].
    psi_deg : float
        Dilation angle [deg].

    Returns
    -------
    apertures : dict
        {
          'step1'    : b1,
          'step2'    : b2,
          'step3'    : b3,
          'delta_bn' : Δb_n,
          'delta_bdil' : Δb_dil
        }
    """
    # Step 1: initial
    b1 = b0

    # Step 2: normal closure from Bandis/Barton law
    delta_bn = normal_closure_hyperbolic(sigma_n, k_n, delta_bm)
    b2 = b0 - delta_bn

    # Step 3: shear-induced dilatancy from slip and dilation angle
    delta_bdil = dilation_from_slip(u_s, psi_deg)
    b3 = b2 + delta_bdil

    return {
        "step1": b1,
        "step2": b2,
        "step3": b3,
        "delta_bn": delta_bn,
        "delta_bdil": delta_bdil,
    }


# ============================================================
# 3. BUILD SURFACES FROM ROUGHNESS + APERTURES
# ============================================================

def build_surfaces_mated(surf_rough, apertures):
    """
    Given a rough lower surface and three apertures, build:

      lower(x,y)        = surf_rough(x,y)
      upper_step_i(x,y) = surf_rough(x,y) + b_step_i
    """
    lower = surf_rough
    upper_surfaces = {
        "step1": surf_rough + apertures["step1"],
        "step2": surf_rough + apertures["step2"],
        "step3": surf_rough + apertures["step3"],
    }
    return lower, upper_surfaces


# ============================================================
# 4. VTK EXPORT
# ============================================================

def export_surface_to_vtk_structured(x, y, surface, filename, scalar_name="height"):
    """
    Export a single surface z(x,y) as a vtkStructuredGrid (.vtk).
    """
    nx, ny = surface.shape
    grid = vtk.vtkStructuredGrid()
    grid.SetDimensions(nx, ny, 1)

    points = vtk.vtkPoints()
    points.SetNumberOfPoints(nx * ny)

    values = vtk.vtkDoubleArray()
    values.SetName(scalar_name)
    values.SetNumberOfValues(nx * ny)

    pid = 0
    for i in range(nx):
        for j in range(ny):
            px = x[i]
            py = y[j]
            pz = surface[i, j]
            points.SetPoint(pid, px, py, pz)
            values.SetValue(pid, surface[i, j])
            pid += 1

    grid.SetPoints(points)
    grid.GetPointData().SetScalars(values)

    writer = vtk.vtkStructuredGridWriter()
    writer.SetFileName(filename)
    writer.SetInputData(grid)
    writer.Write()
    print(f"✅ VTK written: {filename}")



def export_final_surfaces_to_vtk(x, y, lower, upper_surfaces, out_dir="_out"):
    """
    Export ONLY the lower surface and the final (dilated) upper surface to VTK.

    Files:
      lower_surface.vtk
      upper_final.vtk   (== upper_surfaces["step3"])
    """
    os.makedirs(out_dir, exist_ok=True)

    # Lower face
    export_surface_to_vtk_structured(
        x, y, lower,
        os.path.join(out_dir, "lower_surface.vtk"),
        scalar_name="z_lower",
    )

    # Final upper face (after closure + dilatancy): step3
    export_surface_to_vtk_structured(
        x, y, upper_surfaces["step3"],
        os.path.join(out_dir, "upper_final.vtk"),
        scalar_name="z_upper_final",
    )


from mpl_toolkits.mplot3d import Axes3D  # already imported in your other file

def plot_final_surfaces_3d(lower, upper_surfaces, x, y, figsize=(10, 8)):
    """
    Show the lower surface and the final upper surface (step3) in 3D.
    """
    X, Y = np.meshgrid(x, y, indexing="ij")
    Z_low = lower
    Z_up  = upper_surfaces["step3"]

    fig = plt.figure(figsize=figsize)
    ax = fig.add_subplot(111, projection="3d")

    # Lower surface
    ax.plot_surface(
        X, Y, Z_low,
        alpha=0.8, linewidth=0, antialiased=True,
    )

    # Upper surface (final)
    ax.plot_surface(
        X, Y, Z_up,
        alpha=0.6, linewidth=0, antialiased=True,
    )

    ax.set_xlabel("X (m)")
    ax.set_ylabel("Y (m)")
    ax.set_zlabel("Z (m)")
    ax.set_title("Final fracture geometry: lower + upper (step 3)")

    plt.tight_layout()
    plt.show()


# ============================================================
# 5. CROSS-SECTION PLOTS (FACES AT TOP/BOTTOM OF MAGNIFIED GAP)
# ============================================================

def plot_cross_sections_three_stages(lower, upper_surfaces, x, y,
                                     VE=20.0, figsize=(10, 10)):
    """
    Plot a Y-cross-section (central slice) for the three stages,
    with the aperture magnified by VE.
    """
    nx, ny = lower.shape
    j_mid  = ny // 2
    x_line = x
    z_low_y_true = lower[:, j_mid]

    fig, axes = plt.subplots(3, 1, figsize=figsize, sharex=True)

    stage_list = [
        ("Step 1: mated, aperture b₀",               upper_surfaces["step1"]),
        ("Step 2: + normal stress σₙ (closure)",     upper_surfaces["step2"]),
        ("Step 3: + imposed dilation Δb_dil",        upper_surfaces["step3"]),
    ]

    # collect for common y-limits
    all_z_plot = []
    for _, upper_surf in stage_list:
        z_up_y_true = upper_surf[:, j_mid]
        a_true      = z_up_y_true - z_low_y_true
        z_low_plot  = z_low_y_true
        z_up_plot   = z_low_y_true + VE * a_true
        all_z_plot.append(z_low_plot)
        all_z_plot.append(z_up_plot)

    z_min = min(z.min() for z in all_z_plot)
    z_max = max(z.max() for z in all_z_plot)
    pad   = 0.05 * (z_max - z_min)

    for row, (title, upper_surf) in enumerate(stage_list):
        ax = axes[row]

        z_up_y_true = upper_surf[:, j_mid]
        a_true      = z_up_y_true - z_low_y_true
        mean_a      = np.mean(a_true)
        max_a       = np.max(a_true)

        z_low_plot = z_low_y_true
        z_up_plot  = z_low_y_true + VE * a_true

        ax.plot(x_line, z_low_plot, label="Lower surface")
        ax.plot(x_line, z_up_plot,  label="Upper surface (gap ×VE)")
        ax.fill_between(
            x_line, z_low_plot, z_up_plot,
            alpha=0.3,
            label="Aperture (×VE)",
        )

        ax.set_title(title, fontsize=12, pad=8)
        ax.text(
            0.5, 0.93,
            f"Mean b = {mean_a*1e3:.3f} mm, Max b = {max_a*1e3:.3f} mm",
            transform=ax.transAxes,
            ha="center", va="top",
            fontsize=9,
            bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.7),
        )

        ax.set_ylim(z_min - pad, z_max + pad)
        ax.set_ylabel("Elevation (m)\n(aperture × {:.0f})".format(VE))
        ax.grid(True, linestyle="--", alpha=0.4)

        if row == 0:
            ax.legend(loc="upper right", fontsize=8)

    axes[-1].set_xlabel("X (m)")
    fig.suptitle(
        "Fracture evolution: initial → closure → dilation (Y cross-section)\n"
        "(faces at top/bottom of magnified gap)",
        fontsize=16, fontweight="bold", y=0.99,
    )
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()


def export_lower_with_width_to_vtk(x, y, lower, upper_surfaces, out_dir="_out"):
    """
    Export the LOWER surface geometry with a point-based 'width' (aperture)
    field attached:

        width(x,y) = upper_step3(x,y) - lower(x,y)

    VTK:
      - Geometry: z = lower(x,y)
      - Scalars:  'width' (aperture at each point)
    """
    os.makedirs(out_dir, exist_ok=True)

    nx, ny = lower.shape
    X = x
    Y = y

    # final aperture field at step3
    upper_final = upper_surfaces["step3"]
    width_field = upper_final - lower           # [m], point-wise aperture

    grid = vtk.vtkStructuredGrid()
    grid.SetDimensions(nx, ny, 1)

    points = vtk.vtkPoints()
    points.SetNumberOfPoints(nx * ny)

    width_arr = vtk.vtkDoubleArray()
    width_arr.SetName("width")
    width_arr.SetNumberOfValues(nx * ny)

    pid = 0
    for i in range(nx):
        for j in range(ny):
            px = X[i]
            py = Y[j]
            pz = lower[i, j]        # geometry = LOWER surface
            points.SetPoint(pid, px, py, pz)

            width_arr.SetValue(pid, width_field[i, j])  # aperture at this point
            pid += 1

    grid.SetPoints(points)
    grid.GetPointData().AddArray(width_arr)
    grid.GetPointData().SetActiveScalars("width")

    writer = vtk.vtkStructuredGridWriter()
    writer.SetFileName(os.path.join(out_dir, "lower_with_width.vtk"))
    writer.SetInputData(grid)
    writer.Write()
    print("✅ VTK written: lower_with_width.vtk (width projected on lower)")
    
import os
import numpy as np
import pyvista as pv


def make_stretched_z_coords(
    cube_height: float,
    n_layers: int,
    stretch_exponent: float = 1.0,
) -> np.ndarray:
    """
    Build non-uniform z-coordinates for a cube of height `cube_height`.

    - n_layers: number of layers per half (above/below fracture).
      Total layers in z-direction = 2*n_layers + 1.
    - fracture plane is at z = 0 (middle layer).
    - stretch_exponent:
        = 1.0 → almost uniform spacing
        > 1.0 → finer near fracture (z=0), coarser near top/bottom.
    """
    z_half = cube_height / 2.0

    # Normalised coordinate from 0 (fracture) to 1 (top)
    s = np.linspace(0.0, 1.0, n_layers + 1)
    s_stretched = s**stretch_exponent

    # Positive z from 0 to +H/2
    z_pos = z_half * s_stretched
    # Negative z mirrored (no double 0)
    z_neg = -z_pos[1:][::-1]

    # Combine: [-H/2 ... -dz ... 0 ... +dz ... +H/2]
    z_coords = np.concatenate([z_neg, z_pos])
    return z_coords


def create_conformal_cube_with_embedded_fracture(
    surface_vtk_file: str,
    cube_height: float = 1.0,
    n_layers: int = 20,
    stretch_exponent: float = 2.0,
    out_dir: str | None = "_out",
    out_basename: str | None = None,
):
    """
    Create a 3D conformal cube with an embedded rough fracture surface.

    Output:
      - volume_mesh: structured 3D cube (StructuredGrid),
      - fracture_surface: structured 2D fracture (StructuredGrid),
      - combined_mesh: unstructured 3D+2D mesh (UnstructuredGrid) with:
          * 3D hex cells:
              MaterialID = 1 (bottom rock)
              MaterialID = 2 (top rock)
              all fracture fields = 0 (width=0, contact=0, z_mid=0, ...)
          * 2D quad cells (fracture):
              MaterialID = 3
              all fracture fields from original surface as **cell_data**.

    Assumptions:
      - Input is a rectilinear x,y grid with 1 or more z-layers per (x,y).
      - Fracture fields like width/aperture_mid/contact/z_mid are cell-based
        in `surface_mesh.cell_data`.
    """

    print(f"🧊 Creating conformal cube from surface: {surface_vtk_file}")

    if not os.path.exists(surface_vtk_file):
        raise FileNotFoundError(f"Surface VTK file not found: {surface_vtk_file}")

    if out_dir is None:
        out_dir = "_out"
    os.makedirs(out_dir, exist_ok=True)

    if out_basename is None:
        out_basename = os.path.splitext(os.path.basename(surface_vtk_file))[0]

    # -------------------------------------------------------------------------
    # Step 0: Read surface and infer grid structure
    # -------------------------------------------------------------------------
    surface_mesh = pv.read(surface_vtk_file)
    print(f"✅ Loaded surface mesh: {surface_mesh.n_points} points, {surface_mesh.n_cells} cells")

    points = surface_mesh.points
    x = points[:, 0]
    y = points[:, 1]
    z = points[:, 2]

    x_unique = np.unique(x)
    y_unique = np.unique(y)
    nx = len(x_unique)
    ny = len(y_unique)
    print(f"ℹ️ Inferred grid: nx = {nx}, ny = {ny}")

    base_points = nx * ny
    n_points = surface_mesh.n_points

    if n_points % base_points != 0:
        raise ValueError(
            f"Surface points ({n_points}) do not match nx*ny*k "
            f"({base_points} * k). Input is not a regular multi-layer grid."
        )

    nz_layers = n_points // base_points
    print(f"ℹ️ Detected {nz_layers} point-layer(s) per (x,y)")

    # For geometry we only need z distribution along layers
    # Sort points by (x,y,z) so we can reshape into [nx,ny,nz_layers]
    sort_idx = np.lexsort((z, y, x))
    z_sorted = z[sort_idx]
    z_layers = z_sorted.reshape(nx, ny, nz_layers)
    z_eff = np.mean(z_layers, axis=2)  # effective fracture surface z(x,y)

    # -------------------------------------------------------------------------
    # Step 0b: Build fracture cell-based fields (keep cell-based!)
    # -------------------------------------------------------------------------
    # We assume a structured 2D grid has (nx-1)*(ny-1) cells per z-layer.
    cells_per_layer = (nx - 1) * (ny - 1)
    n_cells_total = surface_mesh.n_cells

    fracture_cell_2d: dict[str, np.ndarray] = {}

    if n_cells_total % cells_per_layer != 0:
        print(
            f"⚠️ surface_mesh.n_cells = {n_cells_total}, "
            f"expected multiple of {(nx - 1)*(ny - 1)}. "
            "Cell-based fracture fields will NOT be mapped."
        )
    else:
        n_cell_layers = n_cells_total // cells_per_layer
        print(f"ℹ️ Detected {n_cell_layers} cell-layer(s)")

        for name in surface_mesh.cell_data.keys():
            data = np.array(surface_mesh.cell_data[name])
            if data.ndim == 1:
                # shape: [n_cells_total] → [n_cell_layers, ny-1, nx-1]
                arr_layers = data.reshape(n_cell_layers, ny - 1, nx - 1)
                # average across layers → single [ny-1, nx-1]
                cell2d = np.mean(arr_layers, axis=0)
            elif data.ndim == 2:
                # shape: [n_cells_total, n_comp] → [n_cell_layers, ny-1, nx-1, n_comp]
                n_comp = data.shape[1]
                arr_layers = data.reshape(n_cell_layers, ny - 1, nx - 1, n_comp)
                cell2d = np.mean(arr_layers, axis=0)  # [ny-1, nx-1, n_comp]
            else:
                print(f"⚠️ Skipping cell_data '{name}' with shape {data.shape}")
                continue

            fracture_cell_2d[name] = cell2d
            print(f"   ↳ mapped cell_data '{name}' to 2D shape {cell2d.shape}")

    # -------------------------------------------------------------------------
    # Step 1: Create 3D grid with non-uniform z
    # -------------------------------------------------------------------------
    print("Step 1: Creating 3D grid...")

    nz = n_layers * 2 + 1
    x_coords = np.linspace(x_unique.min(), x_unique.max(), nx)
    y_coords = np.linspace(y_unique.min(), y_unique.max(), ny)

    z_coords = make_stretched_z_coords(
        cube_height=cube_height,
        n_layers=n_layers,
        stretch_exponent=stretch_exponent,
    )

    xx, yy, zz = np.meshgrid(x_coords, y_coords, z_coords, indexing="ij")

    # -------------------------------------------------------------------------
    # Step 2: Embed fracture surface at middle layer, with gradual deformation
    # -------------------------------------------------------------------------
    print("Step 2: Embedding fracture surface...")

    fracture_layer = nz // 2
    fracture_z_level = z_coords[fracture_layer]  # ≈ 0.0

    mean_z_eff = np.mean(z_eff)
    deformation = z_eff - mean_z_eff

    # embed rough surface at middle layer
    zz[:, :, fracture_layer] = fracture_z_level + deformation

    # Gradually decay deformation towards top
    for k in range(fracture_layer + 1, nz):
        blend = (k - fracture_layer) / (nz - 1 - fracture_layer)
        zz[:, :, k] = z_coords[k] + deformation * (1.0 - blend)

    # Bottom half kept flat (can be symmetrised if needed)

    # -------------------------------------------------------------------------
    # Step 3: Structured 3D volume mesh
    # -------------------------------------------------------------------------
    print("Step 3: Creating structured 3D volume mesh...")

    volume_mesh = pv.StructuredGrid(xx, yy, zz)

    # -------------------------------------------------------------------------
    # Step 4: Structured 2D fracture surface (for reference/visualisation)
    # -------------------------------------------------------------------------
    print("Step 4: Building structured 2D fracture surface...")

    fracture_points = np.column_stack(
        [
            xx[:, :, fracture_layer].ravel(),
            yy[:, :, fracture_layer].ravel(),
            zz[:, :, fracture_layer].ravel(),
        ]
    )

    fracture_surface = pv.StructuredGrid()
    fracture_surface.points = fracture_points
    fracture_surface.dimensions = [nx, ny, 1]

    # Add cell_data to 2D structured fracture surface
    n_quads = (nx - 1) * (ny - 1)
    for name, cell2d in fracture_cell_2d.items():
        if cell2d.ndim == 2:
            # [ny-1, nx-1] → [n_quads]
            fracture_surface.cell_data[name] = cell2d.ravel(order="C")
        elif cell2d.ndim == 3:
            # [ny-1, nx-1, n_comp] → [n_quads, n_comp]
            ny2, nx2, n_comp = cell2d.shape
            assert ny2 == ny - 1 and nx2 == nx - 1
            fracture_surface.cell_data[name] = cell2d.reshape(
                n_quads, n_comp, order="C"
            )
        else:
            print(f"⚠️ Skipping 2D cell field '{name}' with shape {cell2d.shape} on fracture surface")

    # You can also tag the 2D fracture itself:
    fracture_surface.cell_data["MaterialID"] = np.full(n_quads, 3, dtype=int)

    # -------------------------------------------------------------------------
    # Step 5: Assign MaterialID for 3D cells (bulk rock)
    # -------------------------------------------------------------------------
    print("Step 5: Assigning MaterialID to 3D cells (rock)...")
    
    n_cells_3d = volume_mesh.n_cells
    # number of cells per z-layer in a structured grid
    n_cells_per_layer = (nx - 1) * (ny - 1)
    
    # sanity check
    assert n_cells_3d == n_cells_per_layer * (nz - 1), (
        f"Mismatch: n_cells_3d={n_cells_3d}, "
        f"(nx-1)*(ny-1)*(nz-1)={n_cells_per_layer*(nz-1)}"
    )
    
    material_ids_3d = np.empty(n_cells_3d, dtype=int)
    
    # VTK/pyvista structured cell ordering: i + (nx-1)*(j + (ny-1)*k)
    for k in range(nz - 1):
        start = k * n_cells_per_layer
        end   = start + n_cells_per_layer
    
        if k < fracture_layer:
            material_ids_3d[start:end] = 1  # bottom rock
        else:
            material_ids_3d[start:end] = 2  # top rock
    
    volume_mesh["MaterialID"] = material_ids_3d


    # -------------------------------------------------------------------------
    # Step 6: Convert structured volume to unstructured hex mesh
    # -------------------------------------------------------------------------
    print("Step 6: Converting volume to unstructured hex mesh...")

    hex_ugrid = volume_mesh.cast_to_unstructured_grid()
    points = hex_ugrid.points.copy()
    cells_hex = hex_ugrid.cells.copy()
    celltypes_hex = hex_ugrid.celltypes.copy()
    n_hex_cells = hex_ugrid.n_cells

    # -------------------------------------------------------------------------
    # Step 7: Build 2D quad cells for fracture (same nodes)
    # -------------------------------------------------------------------------
    print("Step 7: Building 2D quad cells for the fracture (shared nodes)...")

    quad_cells_list = []

    # VTK structured ordering: point index = i + nx * (j + ny * k)
    for j in range(ny - 1):
        for i in range(nx - 1):
            p0 = i + nx * (j + ny * fracture_layer)
            p1 = (i + 1) + nx * (j + ny * fracture_layer)
            p2 = (i + 1) + nx * ((j + 1) + ny * fracture_layer)
            p3 = i + nx * ((j + 1) + ny * fracture_layer)
            quad_cells_list.extend([4, p0, p1, p2, p3])

    cells_quads = np.array(quad_cells_list, dtype=cells_hex.dtype)
    celltypes_quads = np.full(
        n_quads,
        pv.CellType.QUAD,
        dtype=celltypes_hex.dtype,
    )

    # -------------------------------------------------------------------------
    # Step 8: Combine 3D hex + 2D quad into one unstructured mesh
    # -------------------------------------------------------------------------
    print("Step 8: Combining 3D hex and 2D quad cells into one mesh...")

    cells_combined = np.concatenate([cells_hex, cells_quads])
    celltypes_combined = np.concatenate([celltypes_hex, celltypes_quads])

    combined_mesh = pv.UnstructuredGrid(cells_combined, celltypes_combined, points)

    # MaterialID for all cells: bulk (1,2), fracture (3)
    mat_hex = material_ids_3d
    mat_frac = np.full(n_quads, 3, dtype=mat_hex.dtype)
    mat_all = np.concatenate([mat_hex, mat_frac])
    combined_mesh.cell_data["MaterialID"] = mat_all

    # -------------------------------------------------------------------------
    # Step 9: Attach fracture fields as CELL DATA in combined mesh
    # -------------------------------------------------------------------------
    print("Step 9: Mapping fracture cell-based fields into 3D+2D mesh...")

    n_cells_combined = combined_mesh.n_cells

    for name, cell2d in fracture_cell_2d.items():
        if cell2d.ndim == 2:
            # [ny-1, nx-1] → [n_quads]
            frac_vals = cell2d.ravel(order="C")
            # combined array: zeros in bulk, fracture values in 2D cells
            all_vals = np.zeros(n_cells_combined, dtype=float)
            all_vals[n_hex_cells:] = frac_vals
            combined_mesh.cell_data[name] = all_vals

        elif cell2d.ndim == 3:
            # [ny-1, nx-1, n_comp] → [n_quads, n_comp]
            ny2, nx2, n_comp = cell2d.shape
            assert ny2 == ny - 1 and nx2 == nx - 1
            frac_vals = cell2d.reshape(n_quads, n_comp, order="C")
            all_vals = np.zeros((n_cells_combined, n_comp), dtype=float)
            all_vals[n_hex_cells:, :] = frac_vals
            combined_mesh.cell_data[name] = all_vals

        else:
            print(f"⚠️ Skipping 2D cell field '{name}' with shape {cell2d.shape} on combined mesh")

    # Now: width, aperture_mid, contact, z_mid, etc. remain cell-based and
    # live ONLY on the fracture cells (3D bulk cells = 0).

    # -------------------------------------------------------------------------
    # Step 10: Save outputs
    # -------------------------------------------------------------------------
    print("Step 10: Saving output meshes...")

    cube_struct_file = os.path.join(out_dir, f"{out_basename}_cube_structured.vtk")
    frac_struct_file = os.path.join(out_dir, f"{out_basename}_fracture_structured.vtk")
    combined_file = os.path.join(out_dir, f"{out_basename}_cube_with_fracture_unstructured.vtu")

    volume_mesh.save(cube_struct_file)
    fracture_surface.save(frac_struct_file)
    combined_mesh.save(combined_file)

    print(f"✅ Structured cube       : {volume_mesh.n_points} points, {volume_mesh.n_cells} cells")
    print(f"✅ Structured fracture   : {fracture_surface.n_points} points, {fracture_surface.n_cells} cells")
    print(f"✅ Unstructured combined : {combined_mesh.n_points} points, {combined_mesh.n_cells} cells")
    print("💾 Saved:")
    print(f"   - {cube_struct_file}")
    print(f"   - {frac_struct_file}")
    print(f"   - {combined_file}  (3D+2D, MaterialID = 1,2,3 + fracture fields as cell_data)")

    return volume_mesh, fracture_surface, combined_mesh


# ============================================================
# 6. WRAPPER FUNCTION: run everything for given inputs
# ============================================================

def run_xue_style_closure_demo(surf_xue, x, y,
                               b0=1.0e-3,
                               delta_bm=0.8e-3,
                               k_n=5.0e7,
                               sigma_n=5.0e6,
                               u_s=1.143e-3,
                               psi_deg=5.0,
                               VE=20.0,
                               out_dir="_out_xue_closure"):
    """
    Full pipeline:
      - compute apertures b1, b2, b3
      - build mated surfaces
      - export each to VTK
      - plot 3-stage Y-cross-section with magnified aperture
      - build conformal cube with embedded fracture (lower+width)

    u_s and psi_deg control the dilatancy via:

        Δb_dil = u_s · tan(psi_deg)
    """
    os.makedirs(out_dir, exist_ok=True)

    apertures = compute_aperture_stages(
        b0, delta_bm, k_n, sigma_n, u_s, psi_deg
    )

    print("Mean mechanical apertures:")
    print(f"  Step 1: b1 = {apertures['step1']*1e3:.3f} mm (initial)")
    print(f"  Step 2: b2 = {apertures['step2']*1e3:.3f} mm (after σₙ)")
    print(f"  Step 3: b3 = {apertures['step3']*1e3:.3f} mm (after dilation)")
    print(f"  Δb_n   = {apertures['delta_bn']*1e3:.3f} mm (normal closure)")
    print(f"  Δb_dil = {apertures['delta_bdil']*1e3:.3f} mm (shear dilatancy)")

    # Build lower + upper surfaces
    lower, upper_surfaces = build_surfaces_mated(surf_xue, apertures)

    # Export lower+upper surfaces (your original 2×VTK)
    export_final_surfaces_to_vtk(x, y, lower, upper_surfaces, out_dir=out_dir)

    # Export lower surface with width field for the cube builder
    export_lower_with_width_to_vtk(x, y, lower, upper_surfaces, out_dir=out_dir)

    # Visualisation
    plot_final_surfaces_3d(lower, upper_surfaces, x, y, figsize=(10, 8))
    plot_cross_sections_three_stages(lower, upper_surfaces, x, y, VE=VE)

    # Path to the surface we just wrote
    surface_vtk_path = os.path.join(out_dir, "lower_with_width.vtk")

    volume_mesh, fracture_surface, combined_mesh = create_conformal_cube_with_embedded_fracture(
        surface_vtk_file=surface_vtk_path,
        cube_height=1.0,
        n_layers=3,
        stretch_exponent=2.0,
        out_dir=out_dir,
        out_basename="lower_with_width",
    )

    return volume_mesh, fracture_surface, combined_mesh


# ============================================================
# 7. CALL
# ============================================================

volume_mesh, fracture_surface, combined_mesh = run_xue_style_closure_demo(
    surf_xue, x, y,
    b0=1.0e-3,
    delta_bm=0.8e-3,
    k_n=5.0e7,
    sigma_n=5.0e6,
    # choose u_s and ψ:
    # here u_s ≈ 1.143e-3 m with ψ = 5° gives Δb_dil ≈ 0.1 mm
    u_s=1.143e-3,
    psi_deg=5.0,
    VE=20.0,
    out_dir="_out_xue_closure"
)
```
