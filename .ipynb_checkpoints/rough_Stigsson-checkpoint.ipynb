
## 1. Self-affine synthetic fracture surfaces (Stigsson-type, FFT)

Fracture surfaces are represented as **self-affine mono-fractal** surfaces, following the approach of Stigsson and Mas Ivars (2018) and Stigsson *et al.* (2025).
The fracture geometry is described by a height (or aperture) function $a(x,y)$ defined over a planar domain; $a(x,y)$ is the elevation of the surface above a reference plane. In the implementation, $a(x,y)$ and $z(x,y)$ are used interchangeably as the surface height.

A self-affine surface is characterized by:

1. The **Hurst exponent** $H$, which controls how roughness grows with scale. For a self-affine surface embedded in 3D,
   $$
   H = 3 - D_\mathrm{surface},
   $$
   where $D_\mathrm{surface}$ is the fractal dimension of the surface. A smooth plane has $D_\mathrm{surface} = 2$ and $H = 1$; rougher surfaces have $D_\mathrm{surface} \to 3$ and thus smaller $H$.

2. A **magnitude parameter** for the height differences, usually taken as the standard deviation of height differences $\sigma_{\delta h}(\Delta \ell)$ at a given lag $\Delta \ell$.

For a one-dimensional trace $h(x)$ of the surface, the height difference over a step $\Delta \ell$ is
$$
\delta h(x;\Delta \ell) = h(x + \Delta \ell) - h(x),
$$

and the corresponding standard deviation is

$$
\sigma_{\delta h}(\Delta \ell)
=\sqrt{\big\langle \big[h(x+\Delta \ell) - h(x)\big]^2 \big\rangle},
$$
where $\langle\cdot\rangle$ denotes averaging along the profile. Following Stigsson *et al.*, the reference lag is
$$
\Delta \ell = 1,\mathrm{mm}
\quad\Rightarrow\quad
\sigma_{\delta h}(1,\mathrm{mm}).
$$

### 1.1 JRC–$H$–$\sigma_{\delta h}$ link

Stigsson and Mas Ivars link the fractal parameters $(H, \sigma_{\delta h})$ to the classical Joint Roughness Coefficient (JRC) by fitting a multiple linear model to Barton’s ten standard roughness profiles. Based on those data they obtain

$$ \mathrm{JRC}=
-4.3 * 54.6,\sigma_{\delta h}(1,\mathrm{mm})
* 4.3,H,$$


  where $\sigma_{\delta h}(1,\mathrm{mm})$ is expressed in meters. This provides a continuous mapping between an engineering JRC value and the pair $(H, \sigma_{\delta h})$.

Self-affine scaling implies
$$
\sigma_{\delta h}(\Delta \ell) \propto \Delta \ell^{H},
$$
so for a grid spacing $dx$ (in meters):
$$
\sigma_{\delta h}(1,\mathrm{mm})
= \sigma_{\delta h}(dx),\Big(\tfrac{1,\mathrm{mm}}{dx}\Big)^{H},
\qquad
\sigma_{\delta h}(dx)
= \sigma_{\delta h}(1,\mathrm{mm})
\Big(\tfrac{dx}{1,\mathrm{mm}}\Big)^{H}.
$$
These relations allow conversion between roughness measured at a given sampling step $dx$ and the value required in the JRC formula.

In the present implementation, this **continuous** JRC–$H$–$\sigma_{\delta h}$ relation is *not* recomputed. Instead, a **fixed table** is used for three JRC values taken directly from Stigsson’s parameter study:

* $\mathrm{JRC} = 4, 7, 10$,
* corresponding $H = 0.7, 0.8, 0.9$,
* corresponding $\sigma_{\delta h}(1,\mathrm{mm}) = 0.0969, 0.1440, 0.1910\ \mathrm{mm}$.

**Reason:** using the published table is simpler and guarantees exact consistency with Stigsson *et al.* for the studied JRC values, without re-implementing the regression.

### 1.2 Power spectrum and FFT-based generation

To generate synthetic self-affine surfaces with prescribed $H$ and $\sigma_{\delta h}$, Stigsson *et al.* use a **spectral (FFT-based)** method. The (one-dimensional) power spectral density is assumed to follow a power law
$$
S(f) = c_I, f^{-\beta_\mathrm{ps}},
$$
with spectral slope
$$
\beta_\mathrm{ps} = 2H + 1,
$$
which is the standard relation for self-affine profiles. The intercept $c_I$ is chosen such that the resulting surface has the target $\sigma_{\delta h}(1,\mathrm{mm})$.

In the present work, the same idea is implemented directly in **2D**:

1. A periodic grid of size $2049 \times 2049$ points is defined, with
   $$
   \Delta x = \Delta y = 1,\mathrm{mm},
   $$
   giving a domain of about $2.049 \times 2.049,\mathrm{m}^2$.

2. FFT frequencies $(k_x, k_y)$ are computed in both directions and the radial wavenumber
   $$
   K = \sqrt{k_x^2 + k_y^2}
   $$
   is formed.

3. The spectral amplitude is prescribed as
   $$
   A(K) \propto K^{-\beta_\mathrm{ps}/2}, \qquad A(0) = 0,
   $$
   and a **band-limit** is applied: only modes with
   $$
   K_\mathrm{min} \leq K \leq K_\mathrm{max}
   $$
   are kept. Here $K_\mathrm{min}$ and $K_\mathrm{max}$ are determined from

   * the grid fundamental and Nyquist frequencies, and
   * optional user-specified smallest and largest wavelengths $\lambda_\mathrm{min},\lambda_\mathrm{max}$.

   Stigsson *et al.* use the grid-implied band; introducing an explicit $[\lambda_\mathrm{min},\lambda_\mathrm{max}]$ is a controlled simplification.

4. Independent random phases $\theta(\mathbf{k}) \sim \mathcal{U}(0,2\pi)$ are drawn and the complex spectrum is built as
   $$
   \hat z(\mathbf{k})
   = A(K)\bigl(\cos\theta(\mathbf{k}) + i\sin\theta(\mathbf{k})\bigr).
   $$

5. A 2D inverse FFT yields a real surface
   $$
   z_\mathrm{big}(x,y) = \Re{\mathrm{IFFT}[\hat z]},
   $$
   which is then shifted to zero mean.

This is the same conceptual spectral method as in Stigsson and Mas Ivars (power-law spectrum + random phase), applied in two spatial dimensions.

### 1.3 Numerical calibration to $\sigma_{\delta h}(1,\mathrm{mm})$

In the original formulation, the spectral intercept $c_I$ is derived analytically so that the generated surface has the correct $\sigma_{\delta h}(1,\mathrm{mm})$.

Here, the intercept is **not** computed analytically. Instead, the amplitude is calibrated **numerically** on the discrete grid:

* The standard deviation of adjacent height differences is computed as
  $$
  \sigma_{\delta h,\text{num}}
  = \mathrm{std}\Bigl(z_{i+1,j} - z_{i,j},; z_{i,j+1} - z_{i,j}\Bigr),
  $$
  using all horizontal and vertical neighbour pairs.

* The surface is rescaled as
  $$
  z_\mathrm{big}
  ;\leftarrow;
  z_\mathrm{big},
  \frac{\sigma_{\delta h,\text{target}}}{\sigma_{\delta h,\text{num}}},
  $$
  where $\sigma_{\delta h,\text{target}}$ is the value implied by the chosen JRC and $H$ (through the Stigsson table).

**Reason:** this avoids reproducing the somewhat involved analytic expression for $c_I$ and enforces the target $\sigma_{\delta h}(1,\mathrm{mm})$ *exactly* on the discrete grid. The statistical outcome is equivalent to the original method within numerical accuracy.

### 1.4 Extraction and mated walls

Following Stigsson *et al.*, only the interior of the large periodic surface is used to limit edge artefacts from FFT periodicity:

* A central $1001 \times 1001$ window is extracted from the $2049 \times 2049$ field, corresponding to $1.0 \times 1.0,\mathrm{m}^2$ at $1,\mathrm{mm}$ resolution.

* The mean is removed:
  $$
  z_0(x,y) = z(x,y) - \langle z\rangle.
  $$

* Two **perfectly mated** fracture walls are then defined as
  $$
  z_\mathrm{lower}(x,y) = -\tfrac{1}{2} z_0(x,y), \qquad
  z_\mathrm{upper}(x,y) = +\tfrac{1}{2} z_0(x,y).
  $$

In this way, the initial joint consists of two statistically self-affine, exactly matching surfaces, consistent with the assumptions in Stigsson *et al.*, but implemented explicitly on a structured grid.

## 2. Shear damage (semi-analytical model of Casagrande *et al.*)

The shear damage algorithm is based on the semi-analytical model proposed by Casagrande *et al.* (2018).

The original method:

* discretises the joint into triangular facets,
* computes a local apparent dip $\beta_{\mathrm{app},i}$ for each facet,
* evaluates shear-breakage and sliding forces on sets of facets with $\beta_{\mathrm{app},i} \geq \beta^*$,
* iterates over decreasing thresholds $\beta^*$.

The implementation follows this **logic** but uses regular grid cells instead of triangles.

### 2.1 Apparent dip

For shear in $x$, the apparent dip per cell is computed as

$$
\beta_{fs}
= \arctan\left(
\frac{z_{i+1,j} + z_{i+1,j+1} - z_{i,j} - z_{i,j+1}}{2,\Delta x}
\right),
$$

(and an analogous expression for shear in $y$).

The function `compute_apparent_dip_signed` returns the signed $\beta_{fs}$ in degrees for each cell.

### 2.2 Shear and sliding forces

Casagrande *et al.* define two macroscopic forces over a set of contributing facets:

* shear-breakage resistance
  $$
  F_\mathrm{shear}
  = \sum_{cf} a_{cf}
  \left(c_\mathrm{joint} + \sigma_{N,cf} \tan\varphi\right),
  $$
* sliding resistance
  $$
  F_\mathrm{slide}
  = \sum_{cf} a_{cf},
  \sigma_{N,cf},\tan\bigl(\varphi_b + \beta_{fs}\bigr),
  $$

with cell area $a_{cf} = \Delta x,\Delta y$, joint cohesion $c_\mathrm{joint}$, friction angle $\varphi$, basic friction angle $\varphi_b$ and local normal stress $\sigma_{N,cf}$.

In the present model:

* facets are grid cells,
* only facets with $\beta_{fs} > 0$ (facing shear) contribute to $F_\mathrm{slide}$,
* the local normal stress is approximated as in Stigsson *et al.*:
  $$
  \sigma_{N,cf}
  = \min\left(\frac{\sigma_n}{\cos\beta_{fs}},\ \mathrm{JCS}\right),
  $$
  with JCS from a stress-dependent table (Section 3).

This is implemented in `calculate_forces_facetwise`.

### 2.3 Facet activation and iteration over $\beta^*$

As in Casagrande *et al.*, facets are activated by a threshold $\beta^*$:

1. for a given $\beta^*$, all cells with $|\beta_{fs}| \geq \beta^*$ are marked as active;
2. $F_\mathrm{shear}$ and $F_\mathrm{slide}$ are computed over active cells;
3. if $F_\mathrm{slide} \geq F_\mathrm{shear}$, these cells are considered to shear (geometry is altered); otherwise $\beta^*$ is reduced.

The function `casagrande_shear_v3` starts from the current maximum apparent dip $\beta_\mathrm{max}$ and decreases $\beta^*$ in steps of $0.5^\circ$ until no more shearing occurs.

### 2.4 Geometry update: slope clipping via Poisson solve

Casagrande *et al.* update geometry by enforcing shear along a plane at $\beta^*$ and then correcting steps; this is formulated for triangular facets.

The implementation replaces this with a regular-grid procedure:

1. compute node-based gradients
   $$
   g_x = \frac{\partial z}{\partial x},\quad
   g_y = \frac{\partial z}{\partial y}
   $$
   with periodic finite differences;
2. map active cells to **active nodes** (nodes belonging to at least one active cell);
3. on active nodes, clip the gradient component in shear direction (e.g. $g_s = g_x$ for shear in $x$) to
   $$
   |g_s| \leq \tan \beta^*,
   $$
   leaving the transverse component unchanged;
4. define the gradient correction $\delta \mathbf{g}$ and solve
   $$
   \nabla^2 \delta z = \nabla\cdot\delta \mathbf{g}
   $$
   in Fourier space with periodic boundary conditions;
5. update the surface: $z \leftarrow z + \delta z$.

**Reason:** this keeps the Casagrande mechanism (reduction of apparent dip in facets that satisfy the force criterion) but is **much simpler to implement** on a regular grid than the original triangular facet scheme. It also avoids artificial large-scale tilts by enforcing gradient consistency via the Poisson solve.

## 3. Stress-dependent friction and JCS (Barton-type)

The stress dependence of friction angle $\varphi$ and JCS follows the Forsmark parameterisation used by Stigsson *et al.* based on Barton’s work.

Instead of continuous functions, the implementation uses a **piecewise table**:

* for $\sigma_n \leq 0.2,\mathrm{MPa}$:
  $$
  \varphi = 60^\circ,\quad \mathrm{JCS} = 209,\mathrm{MPa},
  $$
* for $0.2 < \sigma_n \leq 2,\mathrm{MPa}$:
  $$
  \varphi = 50^\circ,\quad \mathrm{JCS} = 153,\mathrm{MPa},
  $$
* for $\sigma_n > 2,\mathrm{MPa}$:
  $$
  \varphi = 30^\circ,\quad \mathrm{JCS} = 97,\mathrm{MPa}.
  $$

The basic friction angle is fixed at

$$
\varphi_b = 30^\circ,
$$

and the joint cohesion is set to

$$
c_\mathrm{joint} = 1,\mathrm{MPa},
$$

representing a reduced cohesion at the joint interface compared to intact rock (about $28,\mathrm{MPa}$) as reported by Barton and co-authors.

**Reason:** Stigsson *et al.* only study three normal stresses ($0.2$, $2$, $20,\mathrm{MPa}$); a discrete table is sufficient and exactly matches their parameter combinations without re-implementing full Barton strength envelopes.

This is the **Barton part** of the model: JRC controls fractal roughness; $\varphi$, $\varphi_b$, JCS and $c_\mathrm{joint}$ are taken from Barton-type shear strength concepts.

---

## 4. Midplane aperture (Stigsson-style)

After shear damage of the lower wall and translation of the upper wall by one grid cell in the shear direction, the aperture field is computed with the **midplane method** of Stigsson *et al.*

For each cell:

1. centroid heights of the lower and upper surfaces are
   $$
   z_{\ell,c} = \tfrac{1}{4}(z_{\ell,i,j} + z_{\ell,i+1,j} + z_{\ell,i,j+1} + z_{\ell,i+1,j+1}),
   $$
   $$
   z_{u,c} = \tfrac{1}{4}(z_{u,i,j} + z_{u,i+1,j} + z_{u,i,j+1} + z_{u,i+1,j+1}),
   $$
   and
   $$
   \Delta z_c = z_{u,c} - z_{\ell,c}.
   $$

2. local slopes in $x$ and $y$ are computed for each wall, giving $(\partial z/\partial x,\partial z/\partial y)$ for lower and upper surfaces;

3. from these slopes, unit normals $\mathbf{n}_\ell$ and $\mathbf{n}_u$ are built;

4. the **midplane normal** is defined as
   $$
   \mathbf{n}*m = \frac{\mathbf{n}*\ell + \mathbf{n}*u}{\bigl|\mathbf{n}*\ell + \mathbf{n}*u\bigr|},
   $$
   with vertical component $n*{m,z}$;

5. the aperture along the midplane normal is
   $$
   a = \max\left(\frac{\Delta z_c}{n_{m,z}},,0\right).
   $$

Negative values (overlap) are set to zero, representing full contact. The function `compute_midplane_aperture` implements this procedure for all cells.

**Reason:** the midplane method uses the orientations of both walls and gives a more realistic local aperture than a purely vertical distance, especially for rough and sheared surfaces, as argued by Stigsson *et al.*

---

## 5. Normal closure (Barton–Bandis hyperbola)

Normal closure as a function of $\sigma_n$ is represented by a **Bandis–Barton hyperbolic law** (Bandis *et al.*, 1983; Barton and Bandis, 1980).

The Bandis-type relation is

$$
\Delta b_n(\sigma_n)
= \frac{\sigma_n}{\sigma_n + k_n,\Delta b_m},\Delta b_m,
$$

where

* $k_n$ is a normal stiffness,
* $\Delta b_m$ is the maximum possible closure.

In the function `apply_bandis_normal_closure`:

* $b_0 = 1,\mathrm{mm}$ is a reference mechanical aperture,
* $\Delta b_m = 0.8,b_0$,
* $k_n = 6.25\times 10^9,\mathrm{Pa/m}$,
* $\sigma_n$ is the applied normal stress in Pa.

The global closure is then applied **uniformly** to the midplane aperture field:

$$
a_\mathrm{closed}(x,y;\sigma_n)
= \max\bigl(a(x,y;\sigma_n) - \Delta b_n(\sigma_n),,0\bigr).
$$

**Reason:** Bandis-type hyperbolas are standard for joint normal closure and capture the essential nonlinearity observed in experiments. Applying a **single global** $\Delta b_n(\sigma_n)$ keeps the model simple and cheap to compute, while enforcing that mean aperture decreases monotonically with $\sigma_n$. A fully local Barton–Bandis contact law would be more realistic but would require solving a nonlinear normal contact problem; this is intentionally avoided to keep the workflow reproducible and efficient.

This is the explicit **Bandis/Barton** component of the model.

---


Compared to the full formulations by Stigsson and Mas Ivars, Stigsson *et al.*, Casagrande *et al.* and Bandis/Barton, this implementation:

* **keeps** the key physical ideas
  – self-affine roughness with JRC-based $H$ and $\sigma_{\delta h}$,
  – semi-analytical shear damage from force equilibrium of sliding and breakage forces,
  – Barton-type stress-dependent friction and JCS,
  – Bandis-type normal closure;

* **simplifies** aspects that are technically heavy but not central for a parameter study:
  – analytic spectral intercept $c_I$ is replaced by numerical rescaling of $\sigma_{\delta h}(1,\mathrm{mm})$;
  – triangular mesh geometry in Casagrande *et al.* is replaced by a regular grid and a slope-clipping Poisson update;
  – continuous Barton envelopes in $\sigma_n$ are collapsed to a small table of ${\sigma_n,\varphi,\mathrm{JCS}}$ values used by Stigsson *et al.*;
  – local normal closure is approximated by a global hyperbolic shift.




```python
import numpy as np
import math
from numpy.fft import fftfreq, ifft2
import matplotlib as mpl
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# ============================================================
# 0. GLOBAL PARAMETERS (Stigsson-style grid + material props)
# ============================================================

Lx = 1.0   # Length of the surface in x-direction (m)
Ly = 1.0   # Length of the surface in y-direction (m)
NX = 1001  # Number of grid points in x-direction
NY = 1001  # Number of grid points in y-direction
DX = Lx / (NX - 1)  # Grid spacing in x-direction (m)
DY = Ly / (NY - 1)  # Grid spacing in y-direction (m)

# JRC-dependent Hurst exponent and sigma_dh(1 mm) (Stigsson Table 1)
JRC_TABLE = {
    4:  {"H": 0.7, "sigma_dh_mm": 0.0969},
    7:  {"H": 0.8, "sigma_dh_mm": 0.1440},
    10: {"H": 0.9, "sigma_dh_mm": 0.1910},
}

PHI_BASIC_DEG = 30.0          # basic friction angle φ_b [deg]

# Joint cohesion (for shear resistance)
COHESION_INTACT_MPA = 28.0
COHESION_JOINT_MPA  = 1.0
COHESION_JOINT      = COHESION_JOINT_MPA * 1e6  # [Pa]

# Reference initial mechanical aperture for closure
B0_MECH = 1.0e-3  # [m]


# ============================================================
# Matplotlib style (tune for journal)
# ============================================================
mpl.rcParams.update({
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 12,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.dpi": 200,
    "axes.grid": False,
})


# ============================================================
# 0.1 Stress-dependent φ and JCS
# ============================================================

def get_phi_and_jcs_table(sigma_n_mpa: float):
    """
    Piecewise φ and JCS vs σ_n (MPa).
    """
    if sigma_n_mpa <= 0.2 + 1e-9:
        phi_deg = 60.0
        JCS_mpa = 209.0
    elif sigma_n_mpa <= 2.0 + 1e-9:
        phi_deg = 50.0
        JCS_mpa = 153.0
    else:
        phi_deg = 30.0
        JCS_mpa = 97.0
    return np.deg2rad(phi_deg), JCS_mpa * 1e6


# ============================================================
# 1. SURFACE GENERATION
# ============================================================

def _calculate_sigma_dh(surface: np.ndarray) -> float:
    """Std of height differences for adjacent vertices (1 mm lag)."""
    dzdx = surface[1:, :] - surface[:-1, :]
    dzdy = surface[:, 1:] - surface[:, :-1]
    diffs = np.concatenate([dzdx.ravel(), dzdy.ravel()])
    return float(np.std(diffs))


def generate_surface_stigsson(
    jrc: int,
    seed: int | None = None,
    lambda_min: float | None = None,
    lambda_max: float | None = None,
):
    """
    Band-limited self-affine surface with JRC-based H and sigma_dh(1 mm).
    """
    if jrc not in JRC_TABLE:
        raise ValueError(f"JRC {jrc} not in table {list(JRC_TABLE.keys())}")

    H = JRC_TABLE[jrc]["H"]
    sigma_dh_mm = JRC_TABLE[jrc]["sigma_dh_mm"]
    sigma_dh_target = sigma_dh_mm * 1e-3  # [m]

    N_big = 2049   # 2^11 + 1
    dx_big = 1e-3  # 1 mm
    dy_big = 1e-3  # 1 mm

    Lx_big = N_big * dx_big
    Ly_big = N_big * dy_big

    if seed is not None:
        np.random.seed(seed)

    kx = 2.0 * np.pi * fftfreq(N_big, d=dx_big)
    ky = 2.0 * np.pi * fftfreq(N_big, d=dy_big)
    KX, KY = np.meshgrid(kx, ky, indexing="ij")
    K = np.sqrt(KX**2 + KY**2)

    k_min_grid = 2.0 * np.pi / max(Lx_big, Ly_big)
    k_max_grid = np.pi / min(dx_big, dy_big)

    k_min_user = None
    k_max_user = None

    if lambda_max is not None and lambda_max > 0.0:
        k_min_user = 2.0 * np.pi / lambda_max
    if lambda_min is not None and lambda_min > 0.0:
        k_max_user = 2.0 * np.pi / lambda_min

    k_min = k_min_grid if k_min_user is None else max(k_min_grid, k_min_user)
    k_max = k_max_grid if k_max_user is None else min(k_max_grid, k_max_user)

    if k_min >= k_max:
        raise ValueError(
            f"Invalid (lambda_min, lambda_max): empty k-band "
            f"(k_min={k_min:.3e}, k_max={k_max:.3e})."
        )

    beta_ps = 2.0 * H + 1.0
    K_nonzero = K.copy()
    K_nonzero[0, 0] = 1.0

    amplitude = K_nonzero ** (-beta_ps / 2.0)
    amplitude[0, 0] = 0.0

    band_mask = (K >= k_min) & (K <= k_max)
    amplitude *= band_mask

    phase = np.random.uniform(0.0, 2.0 * np.pi, size=(N_big, N_big))
    spectrum = amplitude * (np.cos(phase) + 1j * np.sin(phase))

    z_big = np.real(ifft2(spectrum))
    z_big -= np.mean(z_big)

    sigma_current = _calculate_sigma_dh(z_big)
    if sigma_current <= 0:
        raise RuntimeError("Generated surface has zero roughness.")
    z_big *= sigma_dh_target / sigma_current

    mid = N_big // 2
    half_x = NX // 2
    half_y = NY // 2
    i_start = mid - half_x
    i_end   = i_start + NX
    j_start = mid - half_y
    j_end   = j_start + NY

    z_crop = z_big[i_start:i_end, j_start:j_end]

    x = np.linspace(0.0, Lx, NX)
    y = np.linspace(0.0, Ly, NY)
    return x, y, z_crop


def generate_correlated_surfaces(
    jrc: int,
    seed: int | None = None,
    lambda_min: float | None = None,
    lambda_max: float | None = None,
):
    """
    Two perfectly mated fracture walls from one rough surface.
    """
    x, y, z = generate_surface_stigsson(
        jrc=jrc,
        seed=seed,
        lambda_min=lambda_min,
        lambda_max=lambda_max,
    )

    z0 = z - z.mean()
    z_lower = -0.5 * z0
    z_upper = +0.5 * z0
    return x, y, z_lower, z_upper


# ============================================================
# 2. APPARENT DIP
# ============================================================

def compute_apparent_dip_signed(surface: np.ndarray,
                                dx: float,
                                dy: float,
                                direction: str = "x") -> np.ndarray:
    """
    Signed apparent dip β_fs [deg] per cell in shear direction.
    """
    if direction not in ("x", "y"):
        raise ValueError("direction must be 'x' or 'y'")
    z = surface
    if direction == "x":
        num = (z[1:, :-1] + z[1:, 1:] - z[:-1, :-1] - z[:-1, 1:]) * 0.5
        slope_s = num / dx
    else:
        num = (z[:-1, 1:] + z[1:, 1:] - z[:-1, :-1] - z[1:, :-1]) * 0.5
        slope_s = num / dy

    beta_app = np.arctan(slope_s)
    beta_deg = np.degrees(beta_app)
    return beta_deg


# ============================================================
# 3. FORCES
# ============================================================

def calculate_forces_facetwise(beta_deg: np.ndarray,
                               active_mask: np.ndarray,
                               sigma_n_pa: float,
                               phi: float,
                               phi_b: float,
                               jcs_pa: float,
                               a_cell: float):
    """
    Shear and sliding forces over active facets.
    """
    facing_mask = (beta_deg > 0.0) & active_mask

    if not np.any(facing_mask):
        return 0.0, 0.0

    beta_active = np.deg2rad(beta_deg[facing_mask])

    sigma_N_i = sigma_n_pa / np.cos(beta_active)
    sigma_N_i = np.minimum(sigma_N_i, jcs_pa)

    F_shear_i = a_cell * (COHESION_JOINT + sigma_N_i * math.tan(phi))
    F_slide_i = a_cell * sigma_N_i * np.tan(phi_b + beta_active)

    F_shear = float(np.sum(F_shear_i))
    F_slide = float(np.sum(F_slide_i))
    return F_shear, F_slide


# ============================================================
# 4. SLOPE CLIPPING
# ============================================================

def clip_slope_in_direction(surface: np.ndarray,
                            dx: float,
                            dy: float,
                            direction: str,
                            beta_deg: np.ndarray,
                            active_mask: np.ndarray,
                            b_star_deg: float) -> np.ndarray:
    """
    Incremental slope clipping in shear direction via FFT Poisson solve.
    """
    if direction not in ("x", "y"):
        raise ValueError("direction must be 'x' or 'y'")

    if not np.any(active_mask):
        return surface.copy()

    z = surface.copy()
    nx, ny = z.shape
    tan_b_star = math.tan(math.radians(b_star_deg))

    gx = (np.roll(z, -1, axis=0) - np.roll(z, 1, axis=0)) / (2.0 * dx)
    gy = (np.roll(z, -1, axis=1) - np.roll(z, 1, axis=1)) / (2.0 * dy)

    active_nodes = np.zeros_like(z, dtype=bool)
    active_nodes[:-1, :-1] |= active_mask
    active_nodes[1:,  :-1] |= active_mask
    active_nodes[:-1, 1:]  |= active_mask
    active_nodes[1:,  1:]  |= active_mask

    gx_clip = gx.copy()
    gy_clip = gy.copy()

    if direction == "x":
        g_s = gx
        mask = active_nodes
        gx_clip[mask] = np.sign(g_s[mask]) * np.minimum(
            np.abs(g_s[mask]), tan_b_star
        )
    else:
        g_s = gy
        mask = active_nodes
        gy_clip[mask] = np.sign(g_s[mask]) * np.minimum(
            np.abs(g_s[mask]), tan_b_star
        )

    delta_gx = gx_clip - gx
    delta_gy = gy_clip - gy

    if np.allclose(delta_gx, 0.0) and np.allclose(delta_gy, 0.0):
        return z

    kx = 2.0 * np.pi * np.fft.fftfreq(nx, d=dx)
    ky = 2.0 * np.pi * np.fft.fftfreq(ny, d=dy)
    KX, KY = np.meshgrid(kx, ky, indexing="ij")

    DGx_k = np.fft.fft2(delta_gx)
    DGy_k = np.fft.fft2(delta_gy)

    RHS_k = 1j * KX * DGx_k + 1j * KY * DGy_k

    denom = KX**2 + KY**2
    denom[0, 0] = 1.0

    DeltaZ_k = -RHS_k / denom
    DeltaZ_k[0, 0] = 0.0

    delta_z = np.real(np.fft.ifft2(DeltaZ_k))

    z_new = z + delta_z
    return z_new


# ============================================================
# 5. CASAGRANDE SHEAR DAMAGE
# ============================================================

def casagrande_shear_v3(surface: np.ndarray,
                        sigma_n_mpa: float,
                        dx: float,
                        dy: float,
                        direction: str = "x",
                        b_step_deg: float = 0.5,
                        max_outer: int = 50,
                        verbose: bool = True) -> np.ndarray:
    """
    Casagrande-style shear damage on one wall.
    """
    if direction not in ("x", "y"):
        raise ValueError("direction must be 'x' or 'y'")

    z = surface.copy()

    sigma_n = sigma_n_mpa * 1e6
    phi, jcs = get_phi_and_jcs_table(sigma_n_mpa)
    phi_b = math.radians(PHI_BASIC_DEG)

    A_cell = dx * dy

    if verbose:
        print(f"Casagrande v3: σ_n={sigma_n_mpa:.3f} MPa, φ={math.degrees(phi):.1f}°, "
              f"φ_b={PHI_BASIC_DEG:.1f}°, JCS={jcs/1e6:.1f} MPa")

    for outer in range(max_outer):
        beta_deg = compute_apparent_dip_signed(z, dx, dy, direction)
        beta_flat = beta_deg.ravel()
        beta_max = float(np.max(beta_flat))
        if beta_max < 1e-3:
            if verbose:
                print(f"  Outer {outer}: max β_fs < 0.001°, stopping.")
            break

        if verbose:
            print(f"  Outer {outer}: max β_fs = {beta_max:.3f}°")

        b_star = math.floor(beta_max / b_step_deg) * b_step_deg
        any_sheared_this_outer = False

        while b_star > 0.0:
            active_mask = np.abs(beta_deg) >= b_star
            N_cf = int(active_mask.sum())
            if N_cf == 0:
                b_star -= b_step_deg
                continue

            F_shear, F_slide = calculate_forces_facetwise(
                beta_deg=beta_deg,
                active_mask=active_mask,
                sigma_n_pa=sigma_n,
                phi=phi,
                phi_b=phi_b,
                jcs_pa=jcs,
                a_cell=A_cell,
            )

            if verbose:
                print(f"    b*={b_star:5.2f}°: N_cf={N_cf}, "
                      f"F_shear={F_shear:.3e} N, F_slide={F_slide:.3e} N")

            if F_slide >= F_shear and F_shear > 0.0:
                z_new = clip_slope_in_direction(
                    surface=z,
                    dx=dx,
                    dy=dy,
                    direction=direction,
                    beta_deg=beta_deg,
                    active_mask=active_mask,
                    b_star_deg=b_star,
                )
                if np.allclose(z_new, z):
                    b_star -= b_step_deg
                    continue

                if verbose:
                    delta = np.max(np.abs(z_new - z))
                    print(f"      Geometry updated at b*={b_star:.2f}°, max |Δz|={delta:.3e} m")

                z = z_new
                any_sheared_this_outer = True
                beta_deg = compute_apparent_dip_signed(z, dx, dy, direction)

            else:
                b_star -= b_step_deg

        if not any_sheared_this_outer:
            if verbose:
                print(f"  ✓ No more cells sheared at outer iter {outer}, stopping.")
            break

    return z


# ============================================================
# 6. MIDPLANE APERTURE
# ============================================================

def compute_midplane_aperture(z_lower: np.ndarray,
                              z_upper: np.ndarray,
                              dx: float,
                              dy: float) -> np.ndarray:
    """
    Midplane-based aperture (Stigsson-style).
    """
    if z_lower.shape != z_upper.shape:
        raise ValueError("z_lower and z_upper must have the same shape")

    nx, ny = z_lower.shape
    if nx < 2 or ny < 2:
        raise ValueError("Need at least 2x2 nodes")

    zL_c = 0.25 * (z_lower[:-1, :-1] + z_lower[1:, :-1]
                   + z_lower[:-1, 1:] + z_lower[1:, 1:])
    zU_c = 0.25 * (z_upper[:-1, :-1] + z_upper[1:, :-1]
                   + z_upper[:-1, 1:] + z_upper[1:, 1:])

    dz_c = zU_c - zL_c

    dzdx_L = (z_lower[1:, :-1] + z_lower[1:, 1:]
              - z_lower[:-1, :-1] - z_lower[:-1, 1:]) / (2.0 * dx)
    dzdy_L = (z_lower[:-1, 1:] + z_lower[1:, 1:]
              - z_lower[:-1, :-1] - z_lower[1:, :-1]) / (2.0 * dy)

    dzdx_U = (z_upper[1:, :-1] + z_upper[1:, 1:]
              - z_upper[:-1, :-1] - z_upper[:-1, 1:]) / (2.0 * dx)
    dzdy_U = (z_upper[:-1, 1:] + z_upper[1:, 1:]
              - z_upper[:-1, :-1] - z_upper[1:, :-1]) / (2.0 * dy)

    def normals_from_slopes(dzdx, dzdy):
        nx = -dzdx
        ny = -dzdy
        nz = np.ones_like(dzdx)
        norm = np.sqrt(nx*nx + ny*ny + nz*nz)
        nx /= norm
        ny /= norm
        nz /= norm
        return nx, ny, nz

    nxL, nyL, nzL = normals_from_slopes(dzdx_L, dzdy_L)
    nxU, nyU, nzU = normals_from_slopes(dzdx_U, dzdy_U)

    nxM = nxL + nxU
    nyM = nyL + nyU
    nzM = nzL + nzU

    normM = np.sqrt(nxM*nxM + nyM*nyM + nzM*nzM)
    normM = np.where(normM == 0.0, 1.0, normM)

    nzM /= normM

    eps = 1e-12
    a = dz_c / (nzM + eps)
    a = np.maximum(a, 0.0)
    return a


# ============================================================
# 7. BARTON–BANDIS NORMAL CLOSURE (GLOBAL)
# ============================================================

def apply_bandis_normal_closure(
    a: np.ndarray,
    sigma_n_mpa: float,
    b0: float = B0_MECH,
):
    """
    Global Bandis-type normal closure.
    """
    KN_JOINT = 6.25e9      # [Pa/m]
    CLOSURE_FRAC = 0.8     # Δb_m = 0.8 * b0

    delta_bm = CLOSURE_FRAC * b0
    sigma_n  = sigma_n_mpa * 1e6

    delta_b_n = (sigma_n / (sigma_n + KN_JOINT * delta_bm)) * delta_bm

    a_closed = np.maximum(a - delta_b_n, 0.0)
    return a_closed, float(delta_b_n)


# ============================================================
# Helper: slide upper wall by one cell in +x direction
# ============================================================

def slide_upper_one_cell_x(z):
    z_u = np.roll(z, shift=1, axis=0)
    z_u[0, :] = z_u[1, :]
    return z_u


```

```python
# ============================================================
# MAIN PIPELINE
# ============================================================

if __name__ == "__main__":

    # 1) Generate initial mated surfaces
    jrc = 7
    seed = 123
    x, y, z_lower, z_upper = generate_correlated_surfaces(
        jrc=jrc,
        seed=seed,
        lambda_min=2e-3,  # smallest wavelength ~ 2 mm
        lambda_max=0.5,   # largest wavelength ~ 0.5 m
    )

    x_cells = 0.5 * (x[:-1] + x[1:])
    y_cells = 0.5 * (y[:-1] + y[1:])

    # 2) Run pipeline for multiple normal stresses
    sigmas_mpa = [0.2, 2.0, 20.0]

    z_lower_damaged = {}
    z_upper_slid = {}
    aperture_midplanes = {}

    # stats vs sigma
    stats_mean_raw = {}
    stats_mean_closed = {}
    stats_contact = {}
    stats_delta_bn = {}

    for sigma_n in sigmas_mpa:
        print(f"\n=== σ_n = {sigma_n} MPa ===")
        z_low_dmg = casagrande_shear_v3(
            surface=z_lower,
            sigma_n_mpa=sigma_n,
            dx=DX,
            dy=DY,
            direction="x",
            b_step_deg=0.5,
            max_outer=50,
            verbose=False,
        )
        z_lower_damaged[sigma_n] = z_low_dmg

        z_up_slid = slide_upper_one_cell_x(z_upper)
        z_upper_slid[sigma_n] = z_up_slid

        a_raw = compute_midplane_aperture(z_low_dmg, z_up_slid, DX, DY)
        a_mid, delta_b_n = apply_bandis_normal_closure(a_raw, sigma_n, b0=B0_MECH)

        mean_raw = float(a_raw.mean())
        mean_closed = float(a_mid.mean())
        contact_pct = float(np.mean(a_mid <= 0.0) * 100.0)

        print(f"  mean(a_raw)    = {mean_raw:.3e} m")
        print(f"  Δb_n (Bandis)  = {delta_b_n:.3e} m")
        print(f"  mean(a_closed) = {mean_closed:.3e} m")
        print(f"  contact%       = {contact_pct:6.2f} %")

        aperture_midplanes[sigma_n] = a_mid

        stats_mean_raw[sigma_n] = mean_raw
        stats_mean_closed[sigma_n] = mean_closed
        stats_contact[sigma_n] = contact_pct
        stats_delta_bn[sigma_n] = delta_b_n

    # 3) Global color / z scales
    all_aperture_values = np.concatenate(
        [aperture_midplanes[s].ravel() for s in sigmas_mpa]
    )
    amin = 0.0
    amax = float(np.max(all_aperture_values))
    print(f"Global aperture range: [{amin:.3e}, {amax:.3e}] m")

    # 4) Cross-section definition
    j_mid_cell = (len(y_cells) - 1) // 2
    y_mid_value = y_cells[j_mid_cell]

    Xc, Yc = np.meshgrid(x_cells, y_cells, indexing="ij")

    # 5) Multi-panel figure: rows = stresses, cols = [2D, 3D, cross-section]
    n_sigma = len(sigmas_mpa)
    fig = plt.figure(figsize=(13, 4 * n_sigma))

    gs = mpl.gridspec.GridSpec(
        n_sigma, 3,
        width_ratios=[1.0, 1.1, 0.8],
        wspace=0.35,
        hspace=0.35
    )

    im_for_cbar = None
    stride = 5

    for row, sigma_n in enumerate(sigmas_mpa):
        a_mid = aperture_midplanes[sigma_n]

        ax2d = fig.add_subplot(gs[row, 0])
        im = ax2d.imshow(
            a_mid.T,
            origin="lower",
            extent=(x_cells[0], x_cells[-1], y_cells[0], y_cells[-1]),
            aspect="equal",
            vmin=amin, vmax=amax,
            cmap="viridis",
        )
        if im_for_cbar is None:
            im_for_cbar = im

        ax2d.axhline(y_mid_value, color="white", linestyle="--", linewidth=1.0)

        if row == 0:
            ax2d.set_title("Midplane aperture (2D map)")
        ax2d.set_xlabel("x [m]")
        ax2d.set_ylabel("y [m]")

        ax2d.text(
            0.02, 0.95,
            fr"σₙ = {sigma_n} MPa",
            transform=ax2d.transAxes,
            ha="left", va="top",
            bbox=dict(boxstyle="round", facecolor="white", alpha=0.8, linewidth=0.0),
        )

        ax3d = fig.add_subplot(gs[row, 1], projection="3d")

        X_plot = Xc[::stride, ::stride]
        Y_plot = Yc[::stride, ::stride]
        Z_plot = a_mid[::stride, ::stride]

        surf = ax3d.plot_surface(
            X_plot, Y_plot, Z_plot,
            rstride=1, cstride=1,
            linewidth=0,
            antialiased=True,
            cmap="viridis",
            vmin=amin, vmax=amax,
        )

        aperture_line_3d = a_mid[:, j_mid_cell]
        ax3d.plot(
            x_cells,
            np.full_like(x_cells, y_mid_value),
            aperture_line_3d,
            color="white", linestyle="--",
            linewidth=3,
            label="cross-section"
        )

        ax3d.set_xlim(x_cells[0], x_cells[-1])
        ax3d.set_ylim(y_cells[0], y_cells[-1])
        ax3d.set_zlim(amin, amax)

        ax3d.set_xlabel("x [m]")
        ax3d.set_ylabel("y [m]")
        ax3d.set_zlabel("a [m]")

        ax3d.view_init(elev=35, azim=-135)
        ax3d.tick_params(axis="both", pad=2)
        if row == 0:
            ax3d.set_title("Midplane aperture (3D surface)")

        ax1d = fig.add_subplot(gs[row, 2])

        aperture_line = a_mid[:, j_mid_cell]
        ax1d.plot(x_cells, aperture_line, lw=1.5, color="C0")

        ax1d.set_xlim(x_cells[0], x_cells[-1])
        ax1d.set_ylim(amin, amax)
        ax1d.set_xlabel("x [m]")
        if row == 0:
            ax1d.set_title(fr"Cross-section at $y \approx {y_mid_value:.2f}\,\mathrm{{m}}$")
        ax1d.set_ylabel("a [m]")
        ax1d.grid(True, alpha=0.4)

    cbar_ax = fig.add_axes([0.20, 0.04, 0.60, 0.02])
    cbar = fig.colorbar(im_for_cbar, cax=cbar_ax, orientation="horizontal")
    cbar.set_label("aperture along midplane a [m]")

    fig.suptitle("Midplane-based aperture fields for different normal stresses", y=0.99)
    plt.tight_layout(rect=[0.03, 0.08, 0.97, 0.96])

    # --------------------------------------------------------
    # 6) Second figure: mean aperture & contact vs σ_n
    # --------------------------------------------------------
    fig2, axes = plt.subplots(1, 2, figsize=(10, 4))

    sig_arr = np.array(sigmas_mpa, dtype=float)
    mean_closed_arr = np.array([stats_mean_closed[s] for s in sigmas_mpa])
    contact_arr = np.array([stats_contact[s] for s in sigmas_mpa])

    # left: mean aperture vs σ_n
    ax_mean = axes[0]
    ax_mean.plot(sig_arr, mean_closed_arr, marker="o", lw=1.5)
    ax_mean.set_xlabel(r"normal stress $\sigma_n$ [MPa]")
    ax_mean.set_ylabel(r"mean aperture $\langle a \rangle$ [m]")
    ax_mean.set_title("Mean closed aperture vs normal stress")
    ax_mean.grid(True, alpha=0.4)

    # right: contact area vs σ_n
    ax_c = axes[1]
    ax_c.plot(sig_arr, contact_arr, marker="s", lw=1.5)
    ax_c.set_xlabel(r"normal stress $\sigma_n$ [MPa]")
    ax_c.set_ylabel("contact area [% of cells]")
    ax_c.set_title("Contact area vs normal stress")
    ax_c.set_ylim(0.0, 100.0)
    ax_c.grid(True, alpha=0.4)

    fig2.tight_layout()

    plt.show()
```

```python
# %%
"""
Fracture model verification 
------------------------------------

This runs a series of tests to verify:

1. Midplane aperture implementation on trivial geometries.
2. Roughness generator statistics (sigma_dh and PSD slope).
3. Casagrande-style shear damage (slope clipping) behaviour.
4. Aperture & contact behaviour with rough surfaces, with/without damage.
5. Stress-dependent friction/JCS table (Barton-type).
6. Bandis-type global normal closure (monotonicity vs σ_n).
"""

import numpy as np
import math
import matplotlib.pyplot as plt

# The following functions are expected to be defined/imported above:
# - JRC_TABLE
# - generate_surface_stigsson
# - generate_correlated_surfaces
# - _calculate_sigma_dh
# - casagrande_shear_v3
# - compute_midplane_aperture
# - slide_upper_one_cell_x
# - clip_slope_in_direction
# - get_phi_and_jcs_table
# - apply_bandis_normal_closure


# %%

def summarize_aperture(a: np.ndarray, label: str = ""):
    """Print basic stats for an aperture field a."""
    a_flat = a.ravel()
    mean_a = float(np.mean(a_flat))
    std_a  = float(np.std(a_flat))
    min_a  = float(np.min(a_flat))
    max_a  = float(np.max(a_flat))
    contact_ratio = float(np.mean(a_flat <= 0.0))

    print(f"--- Aperture stats [{label}] ---")
    print(f"  mean(a)   = {mean_a:.3e} m")
    print(f"  std(a)    = {std_a:.3e} m")
    print(f"  min(a)    = {min_a:.3e} m")
    print(f"  max(a)    = {max_a:.3e} m")
    print(f"  contact%  = {100.0*contact_ratio:5.2f} %")

    return {
        "mean": mean_a,
        "std": std_a,
        "min": min_a,
        "max": max_a,
        "contact_ratio": contact_ratio,
    }


# =====================================================================
# STEP 0 – STRESS-DEPENDENT FRICTION / JCS TABLE
# =====================================================================

# %%
def test_phi_jcs_table():
    """
    Check that get_phi_and_jcs_table reproduces the intended piecewise
    (φ, JCS) values at and around the breakpoints 0.2 and 2.0 MPa.
    """
    print("\n=== Testing get_phi_and_jcs_table ===")

    # Exact tabulated points
    for sigma, phi_exp, jcs_exp in [
        (0.2, 60.0, 209.0),
        (2.0, 50.0, 153.0),
        (20.0, 30.0, 97.0),
    ]:
        phi_rad, jcs_pa = get_phi_and_jcs_table(sigma)
        phi_deg = math.degrees(phi_rad)
        jcs_mpa = jcs_pa / 1e6
        print(f"  σ_n={sigma:5.2f} MPa: φ={phi_deg:5.1f}°, JCS={jcs_mpa:6.1f} MPa")

        assert abs(phi_deg - phi_exp) < 1e-9, "φ table mismatch"
        assert abs(jcs_mpa - jcs_exp) < 1e-9, "JCS table mismatch"

    # Points just below/above the breakpoints
    for sigma_test, expected_phi, expected_jcs in [
        (0.1999, 60.0, 209.0),  # below 0.2
        (0.2001, 50.0, 153.0),  # above 0.2
        (1.9999, 50.0, 153.0),  # below 2.0
        (2.0001, 30.0, 97.0),   # above 2.0
    ]:
        phi_rad, jcs_pa = get_phi_and_jcs_table(sigma_test)
        phi_deg = math.degrees(phi_rad)
        jcs_mpa = jcs_pa / 1e6

        assert abs(phi_deg - expected_phi) < 1e-6, f"φ piecewise mismatch at σ_n={sigma_test}"
        assert abs(jcs_mpa - expected_jcs) < 1e-6, f"JCS piecewise mismatch at σ_n={sigma_test}"

    print(">> test_phi_jcs_table: PASSED")


test_phi_jcs_table()


# =====================================================================
# STEP 1 – MIDPLANE APERTURE SANITY TESTS (TRIVIAL GEOMETRIES)
# =====================================================================

# %%
# 1A. Flat, perfectly mated surfaces, no shear
# Expected: uniform aperture = b0, zero std, zero contact.

def test_flat_no_shear(nx=101, ny=101, b0=1e-3):
    dx = dy = 1e-3  # 1 mm (arbitrary for these kinematic tests)

    z_lower = np.zeros((nx, ny))
    z_upper = np.full((nx, ny), b0)

    a = compute_midplane_aperture(z_lower, z_upper, dx, dy)
    stats = summarize_aperture(a, label=f"flat_no_shear (b0={b0} m)")

    assert abs(stats["mean"] - b0) < 1e-8, "Mean aperture deviates from b0"
    assert stats["std"] < 1e-10, "Std of aperture should be ~0 for flat plates"
    assert stats["contact_ratio"] == 0.0, "There should be no contact (a>0)"

    print(">> test_flat_no_shear: PASSED")


test_flat_no_shear()


# %%
# 1B. Flat, perfectly mated surfaces, one-cell shear offset
# Expected: still zero aperture everywhere.

def test_flat_with_shear(nx=101, ny=101):
    dx = dy = 1e-3

    z_lower = np.zeros((nx, ny))
    z_upper = np.zeros((nx, ny))

    z_upper_slid = slide_upper_one_cell_x(z_upper)

    a = compute_midplane_aperture(z_lower, z_upper_slid, dx, dy)
    stats = summarize_aperture(a, label="flat_with_shear")

    assert abs(stats["mean"]) < 1e-12, "Mean aperture should be ~0 for flat mated"
    assert stats["max"] < 1e-12, "Max aperture should be ~0 for flat mated"
    assert stats["contact_ratio"] == 1.0, "Everything should be in contact (a=0)"

    print(">> test_flat_with_shear: PASSED")


test_flat_with_shear()


# =====================================================================
# STEP 2 – ROUGHNESS GENERATOR: sigma_dh AND PSD SLOPE
# =====================================================================

# %%
# 2A. Check sigma_dh(1 mm) calibration for different seeds

def test_sigma_dh_jrc(jrc=7, seeds=range(5)):
    print(f"\n=== Testing sigma_dh for JRC={jrc} ===")
    target_mm = JRC_TABLE[jrc]["sigma_dh_mm"]
    target_m  = target_mm * 1e-3

    vals = []
    for seed in seeds:
        x, y, z = generate_surface_stigsson(jrc=jrc, seed=seed)
        sigma = _calculate_sigma_dh(z)
        vals.append(sigma)
        print(f"  seed={seed:3d}: sigma_dh = {sigma:.4e} m")

    vals = np.array(vals)
    mean_val = float(vals.mean())
    rel_err  = abs(mean_val - target_m) / target_m

    print(f"Target sigma_dh(1 mm) = {target_m:.4e} m")
    print(f"Mean over seeds       = {mean_val:.4e} m")
    print(f"Relative error        = {100.0*rel_err:.2f} %")

    assert rel_err < 0.05, "sigma_dh deviates more than 5% from target on average"
    print(">> test_sigma_dh_jrc: PASSED")


test_sigma_dh_jrc(jrc=7, seeds=range(5))


# %%
# 2B. Check 1D PSD slope from midline traces against expected -(2H)

def estimate_psd_slope_1d(z_line, dx, fmin_frac=0.05, fmax_frac=0.4):
    """
    Estimate PSD slope from a 1D trace using log-log linear regression
    in a given fraction of the frequency range.
    """
    n = len(z_line)
    Z = np.fft.rfft(z_line - z_line.mean())
    freqs = np.fft.rfftfreq(n, d=dx)
    S = (np.abs(Z) ** 2) / n

    mask = freqs > 0.0
    freqs = freqs[mask]
    S = S[mask]

    fmin = freqs[int(fmin_frac * len(freqs))]
    fmax = freqs[int(fmax_frac * len(freqs))]
    band = (freqs >= fmin) & (freqs <= fmax)

    freqs_band = freqs[band]
    S_band = S[band]

    x = np.log10(freqs_band)
    y = np.log10(S_band)

    p = np.polyfit(x, y, 1)
    slope = p[0]
    return slope, (fmin, fmax)


def test_psd_slope(jrc=7, seed=0):
    print(f"\n=== Testing PSD slope for JRC={jrc}, seed={seed} ===")
    H = JRC_TABLE[jrc]["H"]

    # For this generator, 1D slices empirically follow S1(k) ~ k^{-2H}
    expected_slope = -2.0 * H

    x, y, z = generate_surface_stigsson(jrc=jrc, seed=seed)
    dx = x[1] - x[0]

    j_mid = z.shape[1] // 2
    z_line = z[:, j_mid]

    slope, (fmin, fmax) = estimate_psd_slope_1d(z_line, dx)
    print(f"Estimated slope (1D cut) = {slope:.3f}")
    print(f"Expected slope (=-2H)    = {expected_slope:.3f}")
    print(f"Frequency band           = [{fmin:.3e}, {fmax:.3e}]")

    assert abs(slope - expected_slope) < 0.4, "PSD slope too far from expected -2H"
    print(">> test_psd_slope: PASSED")


test_psd_slope(jrc=7, seed=0)


# =====================================================================
# STEP 3 – SHEAR DAMAGE / CLIPPING CHECKS
# =====================================================================

# %%
def compute_max_beta(surface, dx, dy, direction="x"):
    beta_deg = compute_apparent_dip_signed(surface, dx, dy, direction=direction)
    beta_abs_max = float(np.max(np.abs(beta_deg)))
    return beta_abs_max


def test_clipping_beta(jrc=7, sigma_n_mpa=2.0, seed=0):
    """
    Check that apparent dips do not increase after applying
    Casagrande-style damage.
    """
    print(f"\n=== Testing clipping behaviour (beta) for JRC={jrc}, sigma_n={sigma_n_mpa} MPa ===")
    x, y, z_lower, _ = generate_correlated_surfaces(jrc=jrc, seed=seed)
    dx = x[1] - x[0]
    dy = y[1] - y[0]

    beta_before = compute_max_beta(z_lower, dx, dy, direction="x")
    print(f"Max |beta_fs| before damage: {beta_before:.3f} deg")

    z_damaged = casagrande_shear_v3(
        surface=z_lower,
        sigma_n_mpa=sigma_n_mpa,
        dx=dx,
        dy=dy,
        direction="x",
        b_step_deg=0.5,
        max_outer=5,
        verbose=False,
    )

    beta_after = compute_max_beta(z_damaged, dx, dy, direction="x")
    print(f"Max |beta_fs| after damage:  {beta_after:.3f} deg")

    assert beta_after <= beta_before + 1e-6, "Max apparent dip should not increase after clipping"
    print(">> test_clipping_beta: PASSED")


test_clipping_beta(jrc=7, sigma_n_mpa=2.0, seed=0)


# %%
def test_damage_delta_z(jrc=7, sigma_n_mpa=2.0, seed=0):
    """
    Check that damage does not overly distort the surface and
    approximately preserves mean height.
    """
    print(f"\n=== Testing delta z and mean height for damage, JRC={jrc}, sigma_n={sigma_n_mpa} MPa ===")
    x, y, z_lower, _ = generate_correlated_surfaces(jrc=jrc, seed=seed)
    dx = x[1] - x[0]
    dy = y[1] - y[0]

    rough_amp = float(np.max(z_lower) - np.min(z_lower))
    mean_before = float(np.mean(z_lower))

    z_damaged = casagrande_shear_v3(
        surface=z_lower,
        sigma_n_mpa=sigma_n_mpa,
        dx=dx,
        dy=dy,
        direction="x",
        b_step_deg=0.5,
        max_outer=10,
        verbose=False,
    )

    delta_z = z_damaged - z_lower
    max_delta = float(np.max(np.abs(delta_z)))
    mean_after = float(np.mean(z_damaged))

    print(f"Roughness amplitude (initial): {rough_amp:.3e} m")
    print(f"Max |delta z| after damage:    {max_delta:.3e} m")
    print(f"Mean z before = {mean_before:.3e} m, after = {mean_after:.3e} m")

    assert max_delta < 0.5 * rough_amp, "Damage changed surface more than 50% of roughness amplitude"
    assert abs(mean_after - mean_before) < 1e-6, "Mean height should be approximately preserved"

    print(">> test_damage_delta_z: PASSED")


test_damage_delta_z(jrc=7, sigma_n_mpa=2.0, seed=0)


# =====================================================================
# STEP 4 – APERTURE & CONTACT ON ROUGH SURFACES
# =====================================================================

# %%
def test_aperture_no_damage(jrc=7, seed=0):
    """
    Rough, mated walls, one shear step, NO damage.
    Aperture should stay in millimetre scale (no blow-up).
    """
    print(f"\n=== Testing aperture for rough surfaces (no damage), JRC={jrc}, seed={seed} ===")
    x, y, z_lower, z_upper = generate_correlated_surfaces(jrc=jrc, seed=seed)
    dx = x[1] - x[0]
    dy = y[1] - y[0]

    z_upper_slid = slide_upper_one_cell_x(z_upper)
    a = compute_midplane_aperture(z_lower, z_upper_slid, dx, dy)

    stats = summarize_aperture(a, label="rough_no_damage")

    assert stats["max"] < 1e-2, "Max aperture > 1 cm is suspicious for this roughness"
    print(">> test_aperture_no_damage: PASSED")


test_aperture_no_damage(jrc=7, seed=0)


# %%
def test_aperture_with_damage(jrc=7, seed=0, sigmas_mpa=(0.2, 2.0, 20.0)):
    """
    Rough, mated walls, one shear step, WITH damage for several σ_n.
    Aperture should remain in a realistic range.
    """
    print(f"\n=== Testing aperture for rough surfaces WITH damage, JRC={jrc}, seed={seed} ===")
    x, y, z_lower, z_upper = generate_correlated_surfaces(jrc=jrc, seed=seed)
    dx = x[1] - x[0]
    dy = y[1] - y[0]

    for sigma_n in sigmas_mpa:
        print(f"\n--- sigma_n = {sigma_n} MPa ---")

        z_low_dmg = casagrande_shear_v3(
            surface=z_lower,
            sigma_n_mpa=sigma_n,
            dx=dx,
            dy=dy,
            direction="x",
            b_step_deg=0.5,
            max_outer=50,
            verbose=False,
        )

        z_up_slid = slide_upper_one_cell_x(z_upper)
        a = compute_midplane_aperture(z_low_dmg, z_up_slid, dx, dy)
        stats = summarize_aperture(a, label=f"rough_with_damage (sigma_n={sigma_n} MPa)")

        assert stats["max"] < 1e-2, f"Max aperture > 1 cm at sigma_n={sigma_n} MPa is suspicious"

    print(">> test_aperture_with_damage: ALL sigma_n PASSED")


test_aperture_with_damage(jrc=7, seed=0, sigmas_mpa=(0.2, 2.0, 20.0))


# =====================================================================
# STEP 5 – BANDIS-TYPE GLOBAL NORMAL CLOSURE
# =====================================================================

# %%
def test_bandis_closure_monotonic():
    """
    Check that the Bandis-type normal closure Δb_n(σ_n) is monotone
    in σ_n, and that applying closure to a fixed aperture field yields:
      - decreasing mean aperture with σ_n
      - non-decreasing contact ratio with σ_n
    """
    print("\n=== Testing Bandis-type normal closure monotonicity ===")

    # Simple synthetic aperture field: uniform + small perturbation
    nx, ny = 101, 101
    base_a = 1.0e-3  # 1 mm
    rng = np.random.default_rng(123)
    a0 = base_a + 0.1e-3 * rng.standard_normal((nx, ny))
    a0 = np.maximum(a0, 0.0)

    sigmas = [0.2, 2.0, 10.0, 20.0]  # MPa, increasing
    mean_list = []
    contact_list = []
    closures = []

    for sigma in sigmas:
        a_closed, delta_b = apply_bandis_normal_closure(a0, sigma, b0=base_a)
        stats = summarize_aperture(a_closed, label=f"Bandis_closed σ_n={sigma} MPa")
        mean_list.append(stats["mean"])
        contact_list.append(stats["contact_ratio"])
        closures.append(delta_b)
        print(f"  σ_n={sigma:5.2f} MPa: Δb_n = {delta_b:.3e} m")

    # 1) Δb_n must be strictly increasing in σ_n
    for i in range(1, len(sigmas)):
        assert closures[i] > closures[i-1], "Bandis closure Δb_n must increase with σ_n"

    # 2) Mean aperture must decrease with σ_n (allow tiny numerical tolerance)
    for i in range(1, len(sigmas)):
        assert mean_list[i] <= mean_list[i-1] + 1e-12, "Mean aperture should not increase with σ_n"

    # 3) Contact ratio must not decrease
    for i in range(1, len(sigmas)):
        assert contact_list[i] >= contact_list[i-1] - 1e-12, "Contact ratio should not decrease with σ_n"

    print(">> test_bandis_closure_monotonic: PASSED")


test_bandis_closure_monotonic()
```

### Save the surface in VTK format

```python
def write_vtk_joint_structured(
    x,
    y,
    z_lower,
    z_upper,
    a_mid,
    filename,
    contact_tol=1e-12,
):
    """
    Write a 3D structured grid with two layers (lower & upper wall)
    and aperture/contact as CELL_DATA between them.

    VTK DATASET: STRUCTURED_GRID
    DIMENSIONS:  NX x NY x 2
      - k = 0 : lower wall z = z_lower[i,j]
      - k = 1 : upper wall z = z_upper[i,j]

    CELL_DATA is defined on the "prism" between k=0 and k=1:
      - aperture_mid  (from a_mid[i,j])
      - contact       (1 if a_mid <= contact_tol, else 0)
      - z_mid         (midplane centroid height)

    Parameters
    ----------
    x, y : 1D arrays, length NX, NY
        Node coordinates in x and y.
    z_lower, z_upper : 2D arrays, shape (NX, NY)
        Heights of lower and upper walls at nodes.
    a_mid : 2D array, shape (NX-1, NY-1)
        Midplane aperture per cell (output of compute_midplane_aperture).
    filename : str
        Path to output .vtk file.
    contact_tol : float
        Threshold for contact definition (a <= contact_tol ⇒ contact).
    """
    x = np.asarray(x)
    y = np.asarray(y)
    z_lower = np.asarray(z_lower)
    z_upper = np.asarray(z_upper)
    a_mid = np.asarray(a_mid)

    nx = x.size
    ny = y.size
    nz_dim = 2  # lower & upper layer

    if z_lower.shape != (nx, ny):
        raise ValueError(f"z_lower.shape={z_lower.shape} must be (nx, ny)=({nx}, {ny})")
    if z_upper.shape != (nx, ny):
        raise ValueError(f"z_upper.shape={z_upper.shape} must be (nx, ny)=({nx}, {ny})")
    if a_mid.shape != (nx-1, ny-1):
        raise ValueError(f"a_mid.shape={a_mid.shape} must be (nx-1, ny-1)=({nx-1}, {ny-1})")

    n_points = nx * ny * nz_dim
    n_cells = (nx - 1) * (ny - 1) * (nz_dim - 1)  # = (nx-1)*(ny-1)

    # --- compute midplane z at cell centers for convenience ---
    zL_c = 0.25 * (z_lower[:-1, :-1] + z_lower[1:, :-1]
                   + z_lower[:-1, 1:] + z_lower[1:, 1:])
    zU_c = 0.25 * (z_upper[:-1, :-1] + z_upper[1:, :-1]
                   + z_upper[:-1, 1:] + z_upper[1:, 1:])
    z_mid = 0.5 * (zL_c + zU_c)

    # contact flag at cells
    contact_cells = (a_mid <= contact_tol).astype(int)

    with open(filename, "w") as f:
        # ------------------------------------------------------
        # Header
        # ------------------------------------------------------
        f.write("# vtk DataFile Version 3.0\n")
        f.write("Fracture joint: lower+upper+aperture\n")
        f.write("ASCII\n")
        f.write("DATASET STRUCTURED_GRID\n")
        f.write(f"DIMENSIONS {nx} {ny} {nz_dim}\n")
        f.write(f"POINTS {n_points} float\n")

        # ------------------------------------------------------
        # POINTS: order (i fastest, then j, then k)
        # k = 0: lower wall; k = 1: upper wall
        # ------------------------------------------------------
        for k in range(nz_dim):
            for j in range(ny):
                yj = y[j]
                for i in range(nx):
                    xi = x[i]
                    if k == 0:
                        zij = z_lower[i, j]
                    else:
                        zij = z_upper[i, j]
                    f.write(f"{xi:.6e} {yj:.6e} {zij:.6e}\n")

        # ------------------------------------------------------
        # CELL DATA (one cell between lower & upper at each (i,j))
        # cell order: k=0..nz-2, j=0..ny-2, i=0..nx-2 ⇒ here k=0 only
        # ------------------------------------------------------
        f.write(f"\nCELL_DATA {n_cells}\n")

        # Aperture (midplane width)
        f.write("SCALARS aperture_mid float 1\n")
        f.write("LOOKUP_TABLE default\n")
        for j in range(ny - 1):
            for i in range(nx - 1):
                f.write(f"{a_mid[i, j]:.6e}\n")

        # Contact flag (1 = contact, 0 = open)
        f.write("\nSCALARS contact int 1\n")
        f.write("LOOKUP_TABLE default\n")
        for j in range(ny - 1):
            for i in range(nx - 1):
                f.write(f"{int(contact_cells[i, j])}\n")

        # Midplane z (cell-center height)
        f.write("\nSCALARS z_mid float 1\n")
        f.write("LOOKUP_TABLE default\n")
        for j in range(ny - 1):
            for i in range(nx - 1):
                f.write(f"{z_mid[i, j]:.6e}\n")
```

```python
import os
import numpy as np
import pyvista as pv


def make_stretched_z_coords(
    cube_height: float,
    n_layers: int,
    stretch_exponent: float = 1.0,
) -> np.ndarray:
    """
    Build non-uniform z-coordinates for a cube of height `cube_height`.

    - n_layers: number of layers per half (above/below fracture).
      Total layers in z-direction = 2*n_layers + 1.
    - fracture plane is at z = 0 (middle layer).
    - stretch_exponent:
        = 1.0 → almost uniform spacing
        > 1.0 → finer near fracture (z=0), coarser near top/bottom.
    """
    z_half = cube_height / 2.0

    # Normalised coordinate from 0 (fracture) to 1 (top)
    s = np.linspace(0.0, 1.0, n_layers + 1)
    s_stretched = s**stretch_exponent

    # Positive z from 0 to +H/2
    z_pos = z_half * s_stretched
    # Negative z mirrored (no double 0)
    z_neg = -z_pos[1:][::-1]

    # Combine: [-H/2 ... -dz ... 0 ... +dz ... +H/2]
    z_coords = np.concatenate([z_neg, z_pos])
    return z_coords


def create_conformal_cube_with_embedded_fracture(
    surface_vtk_file: str,
    cube_height: float = 1.0,
    n_layers: int = 20,
    stretch_exponent: float = 2.0,
    out_dir: str | None = "_out",
    out_basename: str | None = None,
):
    """
    Create a 3D conformal cube with an embedded rough fracture surface.

    Output:
      - volume_mesh: structured 3D cube (StructuredGrid),
      - fracture_surface: structured 2D fracture (StructuredGrid),
      - combined_mesh: unstructured 3D+2D mesh (UnstructuredGrid) with:
          * 3D hex cells:
              MaterialID = 1 (bottom rock)
              MaterialID = 2 (top rock)
              all fracture fields = 0 (width=0, contact=0, z_mid=0, ...)
          * 2D quad cells (fracture):
              MaterialID = 3
              all fracture fields from original surface as **cell_data**.

    Assumptions:
      - Input is a rectilinear x,y grid with 1 or more z-layers per (x,y).
      - Fracture fields like width/aperture_mid/contact/z_mid are cell-based
        in `surface_mesh.cell_data`.
    """

    print(f"🧊 Creating conformal cube from surface: {surface_vtk_file}")

    if not os.path.exists(surface_vtk_file):
        raise FileNotFoundError(f"Surface VTK file not found: {surface_vtk_file}")

    if out_dir is None:
        out_dir = "_out"
    os.makedirs(out_dir, exist_ok=True)

    if out_basename is None:
        out_basename = os.path.splitext(os.path.basename(surface_vtk_file))[0]

    # -------------------------------------------------------------------------
    # Step 0: Read surface and infer grid structure
    # -------------------------------------------------------------------------
    surface_mesh = pv.read(surface_vtk_file)
    print(f"✅ Loaded surface mesh: {surface_mesh.n_points} points, {surface_mesh.n_cells} cells")

    points = surface_mesh.points
    x = points[:, 0]
    y = points[:, 1]
    z = points[:, 2]

    x_unique = np.unique(x)
    y_unique = np.unique(y)
    nx = len(x_unique)
    ny = len(y_unique)
    print(f"ℹ️ Inferred grid: nx = {nx}, ny = {ny}")

    base_points = nx * ny
    n_points = surface_mesh.n_points

    if n_points % base_points != 0:
        raise ValueError(
            f"Surface points ({n_points}) do not match nx*ny*k "
            f"({base_points} * k). Input is not a regular multi-layer grid."
        )

    nz_layers = n_points // base_points
    print(f"ℹ️ Detected {nz_layers} point-layer(s) per (x,y)")

    # For geometry we only need z distribution along layers
    # Sort points by (x,y,z) so we can reshape into [nx,ny,nz_layers]
    sort_idx = np.lexsort((z, y, x))
    z_sorted = z[sort_idx]
    z_layers = z_sorted.reshape(nx, ny, nz_layers)
    z_eff = np.mean(z_layers, axis=2)  # effective fracture surface z(x,y)

    # -------------------------------------------------------------------------
    # Step 0b: Build fracture cell-based fields (keep cell-based!)
    # -------------------------------------------------------------------------
    # We assume a structured 2D grid has (nx-1)*(ny-1) cells per z-layer.
    cells_per_layer = (nx - 1) * (ny - 1)
    n_cells_total = surface_mesh.n_cells

    fracture_cell_2d: dict[str, np.ndarray] = {}

    if n_cells_total % cells_per_layer != 0:
        print(
            f"⚠️ surface_mesh.n_cells = {n_cells_total}, "
            f"expected multiple of {(nx - 1)*(ny - 1)}. "
            "Cell-based fracture fields will NOT be mapped."
        )
    else:
        n_cell_layers = n_cells_total // cells_per_layer
        print(f"ℹ️ Detected {n_cell_layers} cell-layer(s)")

        for name in surface_mesh.cell_data.keys():
            data = np.array(surface_mesh.cell_data[name])
            if data.ndim == 1:
                # shape: [n_cells_total] → [n_cell_layers, ny-1, nx-1]
                arr_layers = data.reshape(n_cell_layers, ny - 1, nx - 1)
                # average across layers → single [ny-1, nx-1]
                cell2d = np.mean(arr_layers, axis=0)
            elif data.ndim == 2:
                # shape: [n_cells_total, n_comp] → [n_cell_layers, ny-1, nx-1, n_comp]
                n_comp = data.shape[1]
                arr_layers = data.reshape(n_cell_layers, ny - 1, nx - 1, n_comp)
                cell2d = np.mean(arr_layers, axis=0)  # [ny-1, nx-1, n_comp]
            else:
                print(f"⚠️ Skipping cell_data '{name}' with shape {data.shape}")
                continue

            fracture_cell_2d[name] = cell2d
            print(f"   ↳ mapped cell_data '{name}' to 2D shape {cell2d.shape}")

    # -------------------------------------------------------------------------
    # Step 1: Create 3D grid with non-uniform z
    # -------------------------------------------------------------------------
    print("Step 1: Creating 3D grid...")

    nz = n_layers * 2 + 1
    x_coords = np.linspace(x_unique.min(), x_unique.max(), nx)
    y_coords = np.linspace(y_unique.min(), y_unique.max(), ny)

    z_coords = make_stretched_z_coords(
        cube_height=cube_height,
        n_layers=n_layers,
        stretch_exponent=stretch_exponent,
    )

    xx, yy, zz = np.meshgrid(x_coords, y_coords, z_coords, indexing="ij")

    # -------------------------------------------------------------------------
    # Step 2: Embed fracture surface at middle layer, with gradual deformation
    # -------------------------------------------------------------------------
    print("Step 2: Embedding fracture surface...")

    fracture_layer = nz // 2
    fracture_z_level = z_coords[fracture_layer]  # ≈ 0.0

    mean_z_eff = np.mean(z_eff)
    deformation = z_eff - mean_z_eff

    # embed rough surface at middle layer
    zz[:, :, fracture_layer] = fracture_z_level + deformation

    # Gradually decay deformation towards top
    for k in range(fracture_layer + 1, nz):
        blend = (k - fracture_layer) / (nz - 1 - fracture_layer)
        zz[:, :, k] = z_coords[k] + deformation * (1.0 - blend)

    # Bottom half kept flat (can be symmetrised if needed)

    # -------------------------------------------------------------------------
    # Step 3: Structured 3D volume mesh
    # -------------------------------------------------------------------------
    print("Step 3: Creating structured 3D volume mesh...")

    volume_mesh = pv.StructuredGrid(xx, yy, zz)

    # -------------------------------------------------------------------------
    # Step 4: Structured 2D fracture surface (for reference/visualisation)
    # -------------------------------------------------------------------------
    print("Step 4: Building structured 2D fracture surface...")

    fracture_points = np.column_stack(
        [
            xx[:, :, fracture_layer].ravel(),
            yy[:, :, fracture_layer].ravel(),
            zz[:, :, fracture_layer].ravel(),
        ]
    )

    fracture_surface = pv.StructuredGrid()
    fracture_surface.points = fracture_points
    fracture_surface.dimensions = [nx, ny, 1]

    # Add cell_data to 2D structured fracture surface
    n_quads = (nx - 1) * (ny - 1)
    for name, cell2d in fracture_cell_2d.items():
        if cell2d.ndim == 2:
            # [ny-1, nx-1] → [n_quads]
            fracture_surface.cell_data[name] = cell2d.ravel(order="C")
        elif cell2d.ndim == 3:
            # [ny-1, nx-1, n_comp] → [n_quads, n_comp]
            ny2, nx2, n_comp = cell2d.shape
            assert ny2 == ny - 1 and nx2 == nx - 1
            fracture_surface.cell_data[name] = cell2d.reshape(
                n_quads, n_comp, order="C"
            )
        else:
            print(f"⚠️ Skipping 2D cell field '{name}' with shape {cell2d.shape} on fracture surface")

    # You can also tag the 2D fracture itself:
    fracture_surface.cell_data["MaterialID"] = np.full(n_quads, 3, dtype=int)

    # -------------------------------------------------------------------------
    # Step 5: Assign MaterialID for 3D cells (bulk rock)
    # -------------------------------------------------------------------------
    print("Step 5: Assigning MaterialID to 3D cells (rock)...")

    n_cells_3d = volume_mesh.n_cells
    material_ids_3d = np.zeros(n_cells_3d, dtype=int)
    fracture_z_plane = fracture_z_level

    for i in range(n_cells_3d):
        cell = volume_mesh.get_cell(i)
        avg_z = np.mean(cell.points[:, 2])
        material_ids_3d[i] = 2 if avg_z > fracture_z_plane else 1

    volume_mesh["MaterialID"] = material_ids_3d

    # -------------------------------------------------------------------------
    # Step 6: Convert structured volume to unstructured hex mesh
    # -------------------------------------------------------------------------
    print("Step 6: Converting volume to unstructured hex mesh...")

    hex_ugrid = volume_mesh.cast_to_unstructured_grid()
    points = hex_ugrid.points.copy()
    cells_hex = hex_ugrid.cells.copy()
    celltypes_hex = hex_ugrid.celltypes.copy()
    n_hex_cells = hex_ugrid.n_cells

    # -------------------------------------------------------------------------
    # Step 7: Build 2D quad cells for fracture (same nodes)
    # -------------------------------------------------------------------------
    print("Step 7: Building 2D quad cells for the fracture (shared nodes)...")

    quad_cells_list = []

    # VTK structured ordering: point index = i + nx * (j + ny * k)
    for j in range(ny - 1):
        for i in range(nx - 1):
            p0 = i + nx * (j + ny * fracture_layer)
            p1 = (i + 1) + nx * (j + ny * fracture_layer)
            p2 = (i + 1) + nx * ((j + 1) + ny * fracture_layer)
            p3 = i + nx * ((j + 1) + ny * fracture_layer)
            quad_cells_list.extend([4, p0, p1, p2, p3])

    cells_quads = np.array(quad_cells_list, dtype=cells_hex.dtype)
    celltypes_quads = np.full(
        n_quads,
        pv.CellType.QUAD,
        dtype=celltypes_hex.dtype,
    )

    # -------------------------------------------------------------------------
    # Step 8: Combine 3D hex + 2D quad into one unstructured mesh
    # -------------------------------------------------------------------------
    print("Step 8: Combining 3D hex and 2D quad cells into one mesh...")

    cells_combined = np.concatenate([cells_hex, cells_quads])
    celltypes_combined = np.concatenate([celltypes_hex, celltypes_quads])

    combined_mesh = pv.UnstructuredGrid(cells_combined, celltypes_combined, points)

    # MaterialID for all cells: bulk (1,2), fracture (3)
    mat_hex = material_ids_3d
    mat_frac = np.full(n_quads, 3, dtype=mat_hex.dtype)
    mat_all = np.concatenate([mat_hex, mat_frac])
    combined_mesh.cell_data["MaterialID"] = mat_all

    # -------------------------------------------------------------------------
    # Step 9: Attach fracture fields as CELL DATA in combined mesh
    # -------------------------------------------------------------------------
    print("Step 9: Mapping fracture cell-based fields into 3D+2D mesh...")

    n_cells_combined = combined_mesh.n_cells

    for name, cell2d in fracture_cell_2d.items():
        if cell2d.ndim == 2:
            # [ny-1, nx-1] → [n_quads]
            frac_vals = cell2d.ravel(order="C")
            # combined array: zeros in bulk, fracture values in 2D cells
            all_vals = np.zeros(n_cells_combined, dtype=float)
            all_vals[n_hex_cells:] = frac_vals
            combined_mesh.cell_data[name] = all_vals

        elif cell2d.ndim == 3:
            # [ny-1, nx-1, n_comp] → [n_quads, n_comp]
            ny2, nx2, n_comp = cell2d.shape
            assert ny2 == ny - 1 and nx2 == nx - 1
            frac_vals = cell2d.reshape(n_quads, n_comp, order="C")
            all_vals = np.zeros((n_cells_combined, n_comp), dtype=float)
            all_vals[n_hex_cells:, :] = frac_vals
            combined_mesh.cell_data[name] = all_vals

        else:
            print(f"⚠️ Skipping 2D cell field '{name}' with shape {cell2d.shape} on combined mesh")

    # Now: width, aperture_mid, contact, z_mid, etc. remain cell-based and
    # live ONLY on the fracture cells (3D bulk cells = 0).

    # -------------------------------------------------------------------------
    # Step 10: Save outputs
    # -------------------------------------------------------------------------
    print("Step 10: Saving output meshes...")

    cube_struct_file = os.path.join(out_dir, f"{out_basename}_cube_structured.vtk")
    frac_struct_file = os.path.join(out_dir, f"{out_basename}_fracture_structured.vtk")
    combined_file = os.path.join(out_dir, f"{out_basename}_cube_with_fracture_unstructured.vtu")

    volume_mesh.save(cube_struct_file)
    fracture_surface.save(frac_struct_file)
    combined_mesh.save(combined_file)

    print(f"✅ Structured cube       : {volume_mesh.n_points} points, {volume_mesh.n_cells} cells")
    print(f"✅ Structured fracture   : {fracture_surface.n_points} points, {fracture_surface.n_cells} cells")
    print(f"✅ Unstructured combined : {combined_mesh.n_points} points, {combined_mesh.n_cells} cells")
    print("💾 Saved:")
    print(f"   - {cube_struct_file}")
    print(f"   - {frac_struct_file}")
    print(f"   - {combined_file}  (3D+2D, MaterialID = 1,2,3 + fracture fields as cell_data)")

    return volume_mesh, fracture_surface, combined_mesh
```

```python
surface_file = "joint_sigma2MPa.vtk"
out_dir ="_out"
volume_mesh, fracture_surface, combined_mesh = create_conformal_cube_with_embedded_fracture(
    surface_file,
    cube_height=1.0,
    n_layers=3,
    stretch_exponent=2.0,
)
```
